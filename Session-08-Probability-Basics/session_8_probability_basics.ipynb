## Task 1: Basic Probability Function & Coin Toss

**Prompt:** *Write a Python function `calculate_probability(event_count, total_outcomes)` that returns the probability of an event, and use it to find the probability of getting 'Heads' when tossing a coin.*

```python
def calculate_probability(event_count, total_outcomes):
    """
    Calculates the probability of an event.
    Formula: P(E) = Number of favorable outcomes (event_count) / Total possible outcomes
    """
    if total_outcomes <= 0:
        raise ValueError("Total outcomes must be strictly positive (> 0).")
    if event_count < 0:
        raise ValueError("Event count cannot be negative.")
    if event_count > total_outcomes:
        raise ValueError("Event count cannot exceed total possible outcomes.")
        
    return event_count / total_outcomes

# Application: Tossing a fair coin
# Favorable outcome: Getting 'Heads' (1 outcome)
# Total possible outcomes: {'Heads', 'Tails'} (2 outcomes)
heads_event_count = 1
coin_total_outcomes = 2

prob_heads = calculate_probability(heads_event_count, coin_total_outcomes)

print("=" * 60)
print("Task 1: Basic Probability Calculation (Coin Toss)")
print("=" * 60)
print(f"Sample Space S        : {{'Heads', 'Tails'}}")
print(f"Favorable Event E     : {{'Heads'}} (Count = {heads_event_count})")
print(f"Total Outcomes n(S)   : {coin_total_outcomes}")
print(f"Probability P(Heads)  : {prob_heads:.4f} ({prob_heads * 100:.1f}%)")
```

## Task 2: Dice Rolling Simulation (Experimental vs. Theoretical Probability)

**Prompt:** *Simulate rolling a six-sided dice 100 times using Python's random module, count how many times a '6' appears, and calculate the experimental probability of getting a '6'.*  
*Hint: Use `random.randint(1,6)` in a loop and keep a counter for '6'.*

```python
import random

# Setting random seed for reproducible experimental demonstration
random.seed(42)

total_rolls = 100
count_six = 0
all_rolls = []

# Simulate rolling a six-sided dice 100 times
for _ in range(total_rolls):
    roll = random.randint(1, 6)
    all_rolls.append(roll)
    if roll == 6:
        count_six += 1

# Calculate probabilities
experimental_prob = count_six / total_rolls
theoretical_prob = 1 / 6

print("=" * 65)
print("Task 2: Dice Rolling Simulation (100 Rolls)")
print("=" * 65)
print(f"Total Rolls Conducted           : {total_rolls}")
print(f"Number of Times '6' Appeared    : {count_six}")
print(f"Experimental Probability P(6)   : {experimental_prob:.4f} ({experimental_prob * 100:.1f}%)")
print(f"Theoretical Probability P(6)    : {theoretical_prob:.4f} ({theoretical_prob * 100:.2f}%)")
print(f"Difference (Simulation Error)   : {abs(experimental_prob - theoretical_prob):.4f}")

# Distribution of all faces
face_counts = {i: all_rolls.count(i) for i in range(1, 7)}
print(f"\nBreakdown of all faces: {face_counts}")
```

## Task 3: Probability of Selecting a 'Pop' Song from Spotify History

**Prompt:** *Given an array of genres from your last 20 Spotify songs (like `['pop', 'rock', 'pop', 'hiphop', ...]`), write code to calculate the probability of randomly picking a 'pop' song from your recent history.*

## Task 4: Probability of Ordering Both Food and Dessert on Zomato

**Prompt:** *Write a function that, given two arrays — one for users who ordered food on Zomato and one for users who ordered dessert — calculates the probability that a randomly selected user ordered both food and dessert (i.e., intersection over total unique users).*

```python
def calculate_prob_both(food_users, dessert_users):
    """
    Calculates probability that a randomly chosen user ordered BOTH food and dessert.
    Formula: P(Both) = n(Food ∩ Dessert) / n(Food ∪ Dessert)
    """
    set_food = set(food_users)
    set_dessert = set(dessert_users)
    
    # Intersection: users who ordered both
    both_users = set_food.intersection(set_dessert)
    
    # Union: total unique users across both groups
    total_unique_users = set_food.union(set_dessert)
    
    if len(total_unique_users) == 0:
        return 0.0, both_users, total_unique_users
        
    probability = len(both_users) / len(total_unique_users)
    return probability, both_users, total_unique_users

# Sample data: User IDs
food_order_users = ['user1', 'user2', 'user3', 'user4', 'user5', 'user6', 'user7', 'user8']
dessert_order_users = ['user3', 'user4', 'user5', 'user9', 'user10']

prob, both, union = calculate_prob_both(food_order_users, dessert_order_users)

print("=" * 70)
print("Task 4: Zomato Food & Dessert Order Overlap")
print("=" * 70)
print(f"Users who ordered Food (n = {len(food_order_users)})    : {food_order_users}")
print(f"Users who ordered Dessert (n = {len(dessert_order_users)}) : {dessert_order_users}\n")
print(f"Intersection (Ordered Both) n(A ∩ B) : {len(both)} users -> {sorted(both)}")
print(f"Total Unique Users (Union) n(A ∪ B)  : {len(union)} users -> {sorted(union)}")
print(f"\nProbability of ordering both P(Food ∩ Dessert) : {len(both)}/{len(union)} = {prob:.4f} ({prob * 100:.1f}%)")
```

## Task 5: Conditional Probability for Flipkart Recommendation Engine

**Prompt:** *Explain with a short Python code example how conditional probability could be used to improve recommendations in an app like Flipkart (e.g., probability of buying headphones given user bought a phone).*  
*Hint: Use counts of users who bought both items vs total who bought a phone.*

---

### Concept & Mathematical Foundation:
* **Conditional Probability Formula:**
  $$P(\text{Headphones} \mid \text{Phone}) = \frac{P(\text{Phone} \cap \text{Headphones})}{P(\text{Phone})} = \frac{\text{Count}(\text{Bought Both Phone and Headphones})}{\text{Count}(\text{Bought Phone})}$$
* **How E-Commerce Apps Use This:**
  When a user adds a smartphone to their cart, Flipkart's recommender calculates $P(\text{Item } X \mid \text{Phone})$ for thousands of accessories. If $P(\text{Headphones} \mid \text{Phone}) \ge \text{Threshold}$ (e.g., 40%), the system immediately surfaces headphones under *"Frequently Bought Together"* or *"Recommended Accessories"*, dramatically boosting average order value (AOV) and conversion rate.

```python
# Transaction dataset simulation on Flipkart
# 10,000 total customers
total_customers = 10000

# Customers who bought a Phone: 1,200
bought_phone_count = 1200

# Customers who bought BOTH Phone and Headphones: 780
bought_both_count = 780

# Customers who bought Headphones (with or without phone): 1,500
bought_headphones_count = 1500

# Calculate Conditional Probability: P(Headphones | Phone)
# P(B|A) = count(A and B) / count(A)
prob_headphones_given_phone = bought_both_count / bought_phone_count

# General (Unconditional) Probability of buying headphones: P(Headphones)
prob_headphones_general = bought_headphones_count / total_customers

# Recommendation confidence threshold
RECOMMENDATION_THRESHOLD = 0.40  # 40%

print("=" * 70)
print("Task 5: Flipkart Conditional Probability Recommender")
print("=" * 70)
print(f"Total Platform Customers                  : {total_customers:,}")
print(f"Users who bought a Phone n(Phone)         : {bought_phone_count:,}")
print(f"Users who bought Both n(Phone ∩ Headphones): {bought_both_count:,}")
print(f"Unconditional P(Headphones)               : {prob_headphones_general:.2%}")
print(f"\nConditional Probability P(Headphones | Phone) : {bought_both_count} / {bought_phone_count} = {prob_headphones_given_phone:.2%}")

print("\n" + "=" * 70)
print("RECOMMENDER SYSTEM LOGIC & DECISION:")
print("=" * 70)
if prob_headphones_given_phone >= RECOMMENDATION_THRESHOLD:
    print(f"• ACTION: TRIGGER RECOMMENDATION!")
    print(f"• Why: A user who buys a phone is {prob_headphones_given_phone / prob_headphones_general:.1f}x more likely to buy headphones than an average shopper ({prob_headphones_given_phone:.1%} vs {prob_headphones_general:.1%}).")
    print(f"• Implementation: Display 'Frequently Bought Together: Add Wireless Earbuds at 15% off' directly on the phone product & checkout page.")
```
