## Task 2: Merge Workflows & Merge Conflict Resolution

### Problem Statement:
Simulate a team workflow by merging your `feature-playlist` branch into `main` using a merge commit (not fast-forward), then intentionally create a merge conflict by editing the same line in `playlist.md` on both branches, and resolve the conflict using the VS Code merge tool.
*(Hint: Make sure to save different changes to the same line in both main and feature-playlist before merging.)*

### Step-by-Step Workflow:

#### 1. Non-Fast-Forward Merge (`--no-ff`):
```bash
git checkout main
git merge --no-ff feature-playlist -m "Merge branch 'feature-playlist' into main"
```

#### 2. Creating an Intentional Merge Conflict:
- On `main`, edit line 5:
  ```markdown
  3. **Bollywood Dance Party**: High-energy party tracks and dance anthems for weekend workouts.
  ```
  Commit on `main`: `git commit -am "update: replace playlist 3 with Bollywood Dance Party on main"`
- On `feature-playlist`, edit the identical line 5:
  ```markdown
  3. **Retro 90s Classic Rock**: Iconic guitar riffs and vintage rock anthems from legends.
  ```
  Commit on `feature-playlist`: `git commit -am "update: replace playlist 3 with Retro 90s Classic Rock on feature-playlist"`
- Attempt merge:
  ```bash
  git checkout main
  git merge feature-playlist
  # Output: CONFLICT (content): Merge conflict in playlist.md
  ```

#### 3. Conflict Markers Generated:
```markdown
<<<<<<< HEAD
3. **Bollywood Dance Party**: High-energy party tracks and dance anthems for weekend workouts.
=======
3. **Retro 90s Classic Rock**: Iconic guitar riffs and vintage rock anthems from legends.
>>>>>>> feature-playlist
```

#### 4. Resolution with VS Code Merge Tool:
- In the VS Code editor, click **"Accept Both Changes"** (or use the 3-Way Merge Editor).
- Stage and commit the resolved file:
  ```bash
  git add playlist.md
  git commit -m "Merge branch 'feature-playlist' and resolve playlist conflict in playlist.md"
  ```


````python
import subprocess
import os

repo_dir = r"C:\Users\gadhi\.gemini\antigravity-ide\scratch\InstaCloneRepo"

# Graph log showing branching, non-fast-forward merge, and conflict resolution
graph_log = subprocess.run(["git", "-C", repo_dir, "log", "--graph", "--oneline", "-n", "8"], capture_output=True, text=True, check=True).stdout

with open(os.path.join(repo_dir, "playlist.md"), "r", encoding="utf-8") as f:
    resolved_content = f.read()

print("=" * 75)
print("TASK 2: MERGE WORKFLOW & CONFLICT RESOLUTION AUDIT")
print("=" * 75)
print("Git Commit Graph:")
print(graph_log.strip())
print("\n" + "-" * 75)
print("Resolved playlist.md (Post-Merge):")
print("-" * 75)
print(resolved_content.strip())
````

## Task 3: Forking a Public DS Repo & Feature Branching

### Problem Statement:
Fork a public GitHub repository related to data science (for example, a simple dataset repo), clone it to your machine, create a new branch called `feature-zomato-analysis`, and add a file `zomato_analysis.txt` with a one-line description of how you would analyze Zomato order data.

### Terminal Commands:
```bash
# 1. Clone your fork from GitHub
git clone https://github.com/22kathan/datasets-fork.git
cd datasets-fork

# 2. Create and switch to feature-zomato-analysis branch
git checkout -b feature-zomato-analysis

# 3. Create zomato_analysis.txt with analytical specification
echo "Perform multivariate regression on order prep time, transit distance, and weather signals to predict dynamic delivery ETAs and identify delivery delay bottlenecks." > zomato_analysis.txt

# 4. Stage and commit
git add zomato_analysis.txt
git commit -m "feat: add Zomato order data analysis description"
```


````python
import subprocess
import os

fork_dir = r"C:\Users\gadhi\.gemini\antigravity-ide\scratch\datasets-fork"

branch_info = subprocess.run(["git", "-C", fork_dir, "branch", "-v"], capture_output=True, text=True, check=True).stdout
with open(os.path.join(fork_dir, "zomato_analysis.txt"), "r", encoding="utf-8") as f:
    desc_line = f.read().strip()

commit_details = subprocess.run(["git", "-C", fork_dir, "log", "-n", "1", "--stat"], capture_output=True, text=True, check=True).stdout

print("=" * 80)
print("TASK 3: FORK CLONE & FEATURE-ZOMATO-ANALYSIS BRANCH COMMIT")
print("=" * 80)
print(f"Active Branch: {branch_info.strip()}")
print(f"\nOne-Line Analytical Description in 'zomato_analysis.txt':")
print(f"  \"{desc_line}\"")
print("\nCommit Record:")
print(commit_details.strip())
````
