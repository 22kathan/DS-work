## Task 1: Verify Git Installation on Operating System

### Problem Statement:
Install Git on your operating system (Windows, Mac, or Linux), open a terminal, and run `git --version` to confirm the installation.

### Shell Command:

````python
import subprocess

# Run git --version command to verify active installation
git_version_result = subprocess.run(["git", "--version"], capture_output=True, text=True, check=True)

print("=" * 70)
print("TASK 1: CONFIRMING GIT INSTALLATION")
print("=" * 70)
print(f"Detected Git Version: {git_version_result.stdout.strip()}")
print("Operating System    : Windows (x86_64)")
print("-" * 70)
print("Status: Git is properly installed and accessible via PATH.")
````

````python
import subprocess

# Retrieve global git author credentials
username = subprocess.run(["git", "config", "--global", "user.name"], capture_output=True, text=True).stdout.strip()
email = subprocess.run(["git", "config", "--global", "user.email"], capture_output=True, text=True).stdout.strip()

print("=" * 70)
print("TASK 2: GLOBAL GIT USER CONFIGURATION")
print("=" * 70)
print(f"Global Username (user.name)  : {username}")
print(f"Global Email (user.email)     : {email}")
print("-" * 70)
print("Every subsequent commit will attribute authorship to these details.")
````

## Task 3: Initialize Repository & Inspect `.git` Architecture

### Problem Statement:
Create a new folder called `InstaCloneRepo`, initialize it as a Git repository using `git init`, and observe the `.git` folder that gets created.

### Shell Commands:
```bash
mkdir InstaCloneRepo
cd InstaCloneRepo
git init -b main
ls -la .git
```

### Purpose of `.git` Folder Components:
- **`HEAD`**: Points to the currently checked-out branch reference (e.g., `ref: refs/heads/main`).
- **`config`**: Repository-specific settings, remote URLs, and branch configurations.
- **`objects/`**: Object database storing all blobs (file snapshots), trees (directories), and commit objects keyed by SHA-1 hashes.
- **`refs/`**: Stores pointers to commits (heads for local branches, remotes for remote tracking branches, tags).
- **`hooks/`**: Client-side scripts triggered by Git lifecycle events (pre-commit, post-merge, etc.).
- **`description`**: Default description file used by GitWeb.
- **`info/`**: Contains repository-specific exclusion rules (`exclude` file).


````python
import os
import subprocess

repo_path = r"C:\Users\gadhi\.gemini\antigravity-ide\scratch\InstaCloneRepo"
git_dir = os.path.join(repo_path, ".git")

print("=" * 75)
print("TASK 3: INITIALIZE INSTACLONEREPO & OBSERVE .GIT DIRECTORY")
print("=" * 75)
print(f"Repository Path : {repo_path}")
print(f".git Folder Path: {git_dir}")

print("\nInternal Structure of .git Folder:")
print("-" * 75)
for item in os.listdir(git_dir):
    full_p = os.path.join(git_dir, item)
    item_type = "DIR " if os.path.isdir(full_p) else "FILE"
    print(f"  [{item_type}] {item:<15} -> {item}")

with open(os.path.join(git_dir, "HEAD"), "r") as f:
    head_content = f.read().strip()
print(f"\nCurrent HEAD Reference : {head_content}")
````

## Task 4: Clone Public Repository & Interpret `git status`

### Problem Statement:
Clone the public repository `https://github.com/public-apis/public-apis` into your system using `git clone`, then run `git status` inside the cloned folder and describe what you see.

### Shell Commands:
```bash
git clone https://github.com/public-apis/public-apis.git
cd public-apis
git status
```

### Analysis of `git status` Output:
1. **`On branch master`**: Indicates the currently active branch checked out in the working directory is `master`.
2. **`Your branch is up to date with 'origin/master'`**: Confirms our local clone is perfectly synchronized with the remote GitHub repository at `origin/master`. There are no unpushed local commits and no pending unpulled remote changes.
3. **`nothing to commit, working tree clean`**: Verifies that the local working tree exactly matches the HEAD commit snapshot. There are no modified files, no staged changes, and no untracked files.


````python
import subprocess

cloned_repo = r"C:\Users\gadhi\.gemini\antigravity-ide\scratch\public-apis"

status_result = subprocess.run(["git", "-C", cloned_repo, "status"], capture_output=True, text=True, check=True)
branch_result = subprocess.run(["git", "-C", cloned_repo, "branch", "-v"], capture_output=True, text=True, check=True)

print("=" * 75)
print("TASK 4: CLONE PUBLIC-APIS AND ANALYZE GIT STATUS")
print("=" * 75)
print(f"Target Repository : https://github.com/public-apis/public-apis")
print(f"Local Directory   : {cloned_repo}\n")
print("Output of 'git status':")
print("-" * 50)
print(status_result.stdout.strip())
print("-" * 50)

print("\nExplanation of Diagnostics:")
print("1. 'On branch master': The active working branch is 'master'.")
print("2. 'Your branch is up to date with origin/master': Local and remote commit histories match 1-to-1.")
print("3. 'working tree clean': Zero uncommitted, unstaged, or untracked changes exist.")
````

## Task 5: Stage and Commit `playlist.txt` Following Best Practices

### Problem Statement:
Create a file called `playlist.txt` inside your `InstaCloneRepo` folder, add three song names, stage and commit the file using `git add` and `git commit`, and write a meaningful commit message following best practices (e.g., `'Add initial playlist with 3 songs'`).
*(Hint: Make sure your commit message is clear, concise, and describes the change as if you were working on a real music app project.)*

### Conventional Commit Standards:
- Format: `type: concise imperative description`
- Example: `feat: add initial playlist with 3 songs`
- The `feat:` prefix designates a newly added user-facing feature in software and data projects.

### Shell Commands:
```bash
echo "1. Starboy - The Weeknd" > playlist.txt
echo "2. Blinding Lights - The Weeknd" >> playlist.txt
echo "3. Lose Yourself - Eminem" >> playlist.txt
git add playlist.txt
git commit -m "feat: add initial playlist with 3 songs"
```


````python
import subprocess
import os

repo_path = r"C:\Users\gadhi\.gemini\antigravity-ide\scratch\InstaCloneRepo"
playlist_path = os.path.join(repo_path, "playlist.txt")

with open(playlist_path, "r", encoding="utf-8") as f:
    playlist_text = f.read()

log_output = subprocess.run(["git", "-C", repo_path, "log", "--stat", "-n", "1"], capture_output=True, text=True, check=True).stdout

print("=" * 75)
print("TASK 5: STAGING & COMMITTING PLAYLIST.TXT WITH BEST PRACTICES")
print("=" * 75)
print(f"File Path: {playlist_path}\n")
print("Contents of playlist.txt:")
print("-" * 40)
print(playlist_text.strip())
print("-" * 40)

print("\nGit Commit Record & Diff Stat:")
print(log_output.strip())
````
