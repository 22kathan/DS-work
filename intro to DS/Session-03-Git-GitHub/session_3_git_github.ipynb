## Task 1: Verify Git Installation & Configuration

### Problem Statement:
Install Git on your system, open a terminal or command prompt, and run `git --version` to confirm the installation.

### Commands:
```bash
git --version
git config --global user.name "Your Name"
git config --global user.email "your.email@example.com"
```


````python
import subprocess

# Verify Git version using subprocess
git_ver = subprocess.run(["git", "--version"], capture_output=True, text=True, check=True)
user_name = subprocess.run(["git", "config", "user.name"], capture_output=True, text=True)
user_email = subprocess.run(["git", "config", "user.email"], capture_output=True, text=True)

print("=" * 70)
print("TASK 1: GIT INSTALLATION AND IDENTITY VERIFICATION")
print("=" * 70)
print(f"Git Installed Version : {git_ver.stdout.strip()}")
print(f"Configured User Name  : {user_name.stdout.strip()}")
print(f"Configured User Email : {user_email.stdout.strip()}")
print("-" * 70)
print("Git is verified and ready for version control operations.")
````

## Task 5: Avoiding the Notebook Versioning Anti-Pattern with Git

### Problem Statement:
Write a short note (3-4 lines) explaining how using Git and GitHub can help you avoid the problem of having multiple versions of the same Jupyter notebook file (like `analysis_v1.ipynb`, `analysis_final2.ipynb`, etc.) when working on a data science project.

### Core Explanation:
1. **Single Source of Truth**: Git maintains a clean, single filename (e.g., `customer_churn_analysis.ipynb`) while tracking the entire linear evolution of cell code, Markdown, and markdown commentary across unique SHA-1 commit hashes.
2. **Branching & Experimentation**: Data scientists can spawn isolated experimental branches (`git checkout -b feature/xgboost-model`) to test risky algorithms without polluting the verified production baseline on the `main` branch.
3. **Time Travel & Diff Inspection**: If an experiment degrades accuracy, Git allows instantaneous rollback (`git checkout <commit_id>`) and side-by-side diff comparison, permanently eliminating chaotic filename clutter like `final_v2_really_final.ipynb`.
4. **Cloud Collaboration & Peer Code Review**: GitHub pull requests provide structured review checkpoints, ensuring that code changes are peer-verified, reproducible, and safely merged into the master repository.
