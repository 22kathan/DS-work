# 🏏 IPL Batsman Performance & Impact Analysis (`analysis.ipynb`)

This notebook analyzes modern T20 batting dynamics using cleaned, processed metrics. We leverage modular functions from `helper_functions.py` to evaluate run accumulation vs. strike velocity and boundary dependencies.

```python
import pandas as pd
import matplotlib.pyplot as plt
from helper_functions import calculate_strike_rate, calculate_boundary_percentage, categorize_batsman_impact

# Load processed dataset (Constraint: No raw/unprocessed data files)
df = pd.read_csv('processed_data.csv')
print('Imported helper_functions successfully.')
print(f'Processed dataset shape: {df.shape}')
print('Top 5 Batsmen Overview:')
print(df[['Player', 'Team', 'Matches', 'Runs', 'Balls_Faced', 'Strike_Rate', 'Boundary_Pct', 'Impact_Role']].head().to_string())
```

```python
print('--- Aggregated Performance Insights ---')
print(f"Average Runs per Player: {df['Runs'].mean():.2f}")
print(f"Average Strike Rate: {df['Strike_Rate'].mean():.2f}")
print(f"Highest Strike Rate: {df.loc[df['Strike_Rate'].idxmax()]['Player']} ({df['Strike_Rate'].max():.2f})")
print(f"Highest Boundary Reliance: {df.loc[df['Boundary_Pct'].idxmax()]['Player']} ({df['Boundary_Pct'].max():.2f}% of runs from 4s & 6s)")
print('Role Breakdown:')
print(df['Impact_Role'].value_counts())
```

```python
plt.figure(figsize=(10, 6), dpi=150)
colors = {'Elite Match Winner': '#d90429', 'Power Hitter': '#7209b7', 'Anchor': '#0077b6'}

for role, group in df.groupby('Impact_Role'):
    plt.scatter(group['Runs'], group['Strike_Rate'], s=group['Boundary_Pct']*4, 
                label=f"{role} (Bubble: Boundary %)", alpha=0.8, color=colors.get(role, '#2a9d8f'), edgecolors='black', linewidth=1.2)

for _, row in df.iterrows():
    plt.annotate(row['Player'], (row['Runs'] + 6, row['Strike_Rate'] + 0.5), fontsize=8.5, fontweight='bold', alpha=0.85)

plt.axhline(150, color='gray', linestyle='--', alpha=0.6, label='High Strike Rate Threshold (150)')
plt.axvline(500, color='gray', linestyle=':', alpha=0.6, label='Elite 500-Run Mark')

plt.title('IPL Batsman Impact Analysis: Runs vs. Strike Rate & Boundary Dominance', fontsize=13, fontweight='bold', pad=14)
plt.xlabel('Total Runs Scored', fontsize=11, fontweight='semibold')
plt.ylabel('Batting Strike Rate', fontsize=11, fontweight='semibold')
plt.legend(frameon=True, facecolor='#f8f9fa', edgecolor='none', loc='upper left', fontsize=8.5)
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('plot.png')
print('Visualizing Runs vs Strike Rate scatter with boundary percentage scaling...')
print('Visualization successfully exported to plot.png')
```
