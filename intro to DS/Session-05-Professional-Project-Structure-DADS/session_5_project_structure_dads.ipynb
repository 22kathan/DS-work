````python
import os

base_path = "insta-analytics"

def print_tree(startpath, max_depth=3):
    tree_str = f"{os.path.basename(os.path.abspath(startpath))}/\n"
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        if level >= max_depth:
            continue
        indent = '│   ' * level
        subindent = '├── '
        for d in dirs:
            tree_str += f"{indent}{subindent}{d}/\n"
        for f in files:
            tree_str += f"{indent}├── {f}\n"
        break
    return tree_str

print("=" * 75)
print("TASK 1: DATA ANALYTICS PROJECT ARCHITECTURE (INSTA-ANALYTICS)")
print("=" * 75)

for root, dirs, files in os.walk(base_path):
    level = root.replace(base_path, '').count(os.sep)
    indent = ' ' * 4 * level
    folder = os.path.basename(root)
    print(f"{indent}📂 {folder}/")
    subindent = ' ' * 4 * (level + 1)
    for file in files:
        print(f"{subindent}📄 {file}")
````

## Task 3: Git Ignore Configuration (`.gitignore`)

### Problem Statement:
Write a `.gitignore` file for your `insta-analytics` project that ignores all files in `data/raw`, any `.ipynb_checkpoints` folders, and any files ending with `.h5` or `.pkl`.
*(Hint: Use wildcards and folder paths to match these patterns.)*

### Technical Syntax & Rules:
- `data/raw/*`: Matches and ignores every individual file within the `data/raw` path, while preserving the directory boundary.
- `.ipynb_checkpoints/`, `**/.ipynb_checkpoints/`: Recursively excludes hidden checkpoint directories created automatically by Jupyter kernels.
- `*.h5`: Wildcard glob ignoring all Hierarchical Data Format (HDF5) deep learning model weight files.
- `*.pkl`: Wildcard glob ignoring all Python `pickle` and `joblib` serialized binary objects.


````python
gitignore_path = os.path.join("insta-analytics", ".gitignore")

with open(gitignore_path, "r", encoding="utf-8") as f:
    rules = f.read()

print("=" * 75)
print("TASK 3: .GITIGNORE CONFIGURATION FOR INSTA-ANALYTICS")
print("=" * 75)
print(rules.strip())
print("=" * 75)
print("\nVerification of Pattern Matching:")
print("  * data/raw/*           -> Successfully matches all raw dump exports.")
print("  * **/.ipynb_checkpoints/ -> Excludes all auto-generated notebook temp states.")
print("  * *.h5 & *.pkl         -> Excludes large neural net weights & pickle model files.")
````

- **Repository**: [`ageron/handson-ml3`](https://github.com/ageron/handson-ml3)
- **Title**: *Hands-On Machine Learning with Scikit-Learn, Keras, and TensorFlow (3rd Edition)* by Aurélien Géron
- **GitHub Stars**: 40,000+ ⭐

### Comparison with Standard DA/DS Architecture:

#### 1. Three Ways It Follows or Differs:
1. **Differs — Flat Root-Level Notebook Organization**:
   - *Observation*: Rather than segregating analysis notebooks into a dedicated `notebooks/` directory, all 19 chapter notebooks (e.g., `01_the_machine_learning_landscape.ipynb`, `02_end_to_end_machine_learning_project.ipynb`) are dumped directly into the repository root.
   - *Impact*: Crowds the root folder with dozens of multi-megabyte files, obscuring repository documentation and configuration assets.
2. **Differs — Absence of a Reusable Source Code Module (`src/`)**:
   - *Observation*: The repository lacks a standard `src/` or `lib/` package for shared data preprocessors, custom scikit-learn transformers, and plotting functions. Instead, functions are redundantly defined inside notebook code cells.
   - *Impact*: Violates DRY (Don't Repeat Yourself) engineering principles and prevents code from being imported into external Python scripts or microservices.
3. **Follows — Robust Environment & Containerization Standards**:
   - *Observation*: Strictly follows best practices by providing explicit environment reproducibility definitions across multiple tools: `requirements.txt` (pip), `environment.yml` (Conda), `docker/` (Dockerfiles), and `INSTALL.md`.

#### 2. Recommended Organizational Improvement:
- **Proposed Refactoring**:
  1. Relocate all 25+ notebooks into a dedicated directory `notebooks/` organized by chapters: `notebooks/01_ml_fundamentals/`, `notebooks/02_deep_learning/`.
  2. Extract common dataset fetching helpers, custom estimators, and plotting styles into a package `src/handson_utils/` containing `data_loader.py` and `visuals.py` so that notebooks import clean, unit-tested functions:
  ```python
  from src.handson_utils import fetch_california_housing_data, plot_decision_boundary
  ```
