## Task 1: 2x2 Matrix Multiplication (Instagram Engagement)

### Problem Statement:
Given two $2 \times 2$ matrices representing the number of likes and comments on two Instagram posts over two days, write Python code to multiply them and display the resulting matrix.

### Mathematical Intuition:
Let Matrix $A$ represent the engagement counts (Likes and Comments) for **Post 1** and **Post 2**:
$$
A = \begin{bmatrix} 150 & 30 \\ 220 & 45 \end{bmatrix} \quad \begin{matrix} \text{Row 0: Post 1 (150 Likes, 30 Comments)} \\ \text{Row 1: Post 2 (220 Likes, 45 Comments)} \end{matrix}
$$

Let Matrix $B$ represent the weighting / engagement multiplier coefficients across **Day 1** and **Day 2**:
$$
B = \begin{bmatrix} 2 & 3 \\ 1 & 4 \end{bmatrix} \quad \begin{matrix} \text{Row 0: Like Multipliers (Day 1: 2, Day 2: 3)} \\ \text{Row 1: Comment Multipliers (Day 1: 1, Day 2: 4)} \end{matrix}
$$

The matrix product $C = A \times B$ is computed via dot products of rows of $A$ and columns of $B$:
$$
C_{00} = (150 \times 2) + (30 \times 1) = 300 + 30 = 330
$$
$$
C_{01} = (150 \times 3) + (30 \times 4) = 450 + 120 = 570
$$
$$
C_{10} = (220 \times 2) + (45 \times 1) = 440 + 45 = 485
$$
$$
C_{11} = (220 \times 3) + (45 \times 4) = 660 + 180 = 840
$$
$$
C = \begin{bmatrix} 330 & 570 \\ 485 & 840 \end{bmatrix}
$$


```python
import numpy as np

# Define Matrix A: 2x2 matrix representing Likes and Comments for 2 Instagram Posts
# Row 0: Post 1 [Likes, Comments]
# Row 1: Post 2 [Likes, Comments]
matrix_A = np.array([
    [150, 30],
    [220, 45]
])

# Define Matrix B: 2x2 matrix representing conversion/weighting factors across Day 1 and Day 2
# Row 0: Like Multipliers [Day 1, Day 2]
# Row 1: Comment Multipliers [Day 1, Day 2]
matrix_B = np.array([
    [2, 3],
    [1, 4]
])

# Perform matrix multiplication using np.matmul / @ operator
result_matrix = np.matmul(matrix_A, matrix_B)

print("=" * 65)
print("TASK 1: 2x2 MATRIX MULTIPLICATION (INSTAGRAM ENGAGEMENT)")
print("=" * 65)
print("\nMatrix A (Instagram Posts: Likes & Comments):")
print(matrix_A)
print("\nMatrix B (Daily Engagement Multipliers across 2 Days):")
print(matrix_B)
print("\n" + "-" * 65)
print("Resulting Product Matrix C = A x B:")
print(result_matrix)
print("-" * 65)

# Detailed breakdown of dot product steps
print("\nStep-by-Step Dot Product Calculation:")
print(f"  C[0, 0] = (150 * 2) + (30 * 1) = {150*2} + {30*1} = {result_matrix[0, 0]}")
print(f"  C[0, 1] = (150 * 3) + (30 * 4) = {150*3} + {30*4} = {result_matrix[0, 1]}")
print(f"  C[1, 0] = (220 * 2) + (45 * 1) = {220*2} + {45*1} = {result_matrix[1, 0]}")
print(f"  C[1, 1] = (220 * 3) + (45 * 4) = {220*3} + {45*4} = {result_matrix[1, 1]}")
```

## Task 2: Linear Regression Prediction Function ($y = mx + c$)

### Problem Statement:
Create a simple Python function `linear_regression_predict(x, m, c)` that takes an input value `x`, slope `m`, and intercept `c`, and returns the predicted output `y` for a linear regression equation $y = mx + c$.

### Mathematical Intuition:
A simple univariate linear regression model captures the linear relationship between a dependent target variable $y$ and an independent predictor feature $x$:
$$
y = mx + c
$$
Where:
- $x$: The independent input feature / predictor variable.
- $m$: The **slope** (gradient), representing the expected change in $y$ for each unit change in $x$ ($\frac{\Delta y}{\Delta x}$).
- $c$: The **y-intercept**, representing the baseline constant value of $y$ when $x = 0$.
- $y$: The predicted dependent variable / continuous target outcome.


```python
def linear_regression_predict(x, m, c):
    """
    Calculates predicted output y for a linear regression model y = mx + c.
    
    Parameters:
    -----------
    x : float or numpy.ndarray
        The independent variable / input feature (scalar or array).
    m : float
        The slope of the regression line (rate of change of y with respect to x).
    c : float
        The y-intercept (the value of y when x = 0).
        
    Returns:
    --------
    float or numpy.ndarray
        The predicted output value(s) y.
    """
    return m * x + c

print("=" * 65)
print("TASK 2: LINEAR REGRESSION PREDICTION FUNCTION")
print("=" * 65)

# Test with sample parameters: m = 3.5, c = 12.0, x = 4.0
test_x = 4.0
test_m = 3.5
test_c = 12.0
predicted_y = linear_regression_predict(test_x, test_m, test_c)

print(f"Regression Equation : y = {test_m} * x + {test_c}")
print(f"Input Feature (x)   : {test_x}")
print(f"Slope (m)           : {test_m}")
print(f"Intercept (c)       : {test_c}")
print(f"Predicted Output (y): {predicted_y}")
print(f"Verification Check  : {test_m} * {test_x} + {test_c} = {test_m * test_x + test_c}")
```
