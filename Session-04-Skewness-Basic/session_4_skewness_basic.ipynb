## Task 1: Swiggy Daily Order Counts - Skewness & Distribution Plot

**Prompt:** *Given the daily order counts for a Swiggy delivery partner over 10 days: `[12, 15, 17, 14, 13, 16, 200, 18, 14, 15]`, plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.*

```python
import matplotlib.pyplot as plt
import statistics

# Daily order counts for Swiggy delivery partner over 10 days
orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

mean_orders = statistics.mean(orders)
median_orders = statistics.median(orders)
mode_orders = statistics.mode(orders)

print("=" * 65)
print("Task 1: Swiggy Delivery Partner Daily Order Counts")
print("=" * 65)
print(f"Daily Orders Dataset  : {orders}")
print(f"Mean Daily Orders     : {mean_orders:.1f}")
print(f"Median Daily Orders   : {median_orders:.1f}")
print(f"Mode Daily Orders     : {mode_orders:.1f}\n")

# Plotting the daily orders distribution
plt.figure(figsize=(9, 4.5))
days = list(range(1, len(orders) + 1))
plt.plot(days, orders, marker='o', color='#fc8019', linewidth=2.5, markersize=8, label='Daily Orders')
plt.axhline(mean_orders, color='red', linestyle='--', linewidth=2, label=f'Mean = {mean_orders:.1f}')
plt.axhline(median_orders, color='green', linestyle='-', linewidth=2, label=f'Median = {median_orders:.1f}')
plt.title("Swiggy Daily Order Counts Over 10 Days (Right-Skewed Distribution)", fontsize=12, fontweight='bold')
plt.xlabel("Day Number", fontsize=11)
plt.ylabel("Number of Orders Delivered", fontsize=11)
plt.xticks(days)
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig("swiggy_orders_distribution.png", dpi=150)
plt.show()

print("=" * 65)
print("SKEWNESS CLASSIFICATION & REASONING:")
print("=" * 65)
print("• Classification: RIGHT SKEWED (Positively Skewed).")
print("• Reasoning: 9 out of 10 days have normal orders tightly clustered between 12 and 18, but Day 7 contains an extreme outlier of 200 orders.")
print("• This extreme high value creates a long tail stretching far out to the right side.")
print("• Consequently, the Mean (32.4) is pulled significantly higher than the Median (15.0), confirming a right-skewed shape.")
```

## Task 2: Instagram Influencer Followers - Mean vs. Median & Skewness

**Prompt:** *Take the following list of Instagram influencer follower counts (in thousands): `[5, 7, 8, 8, 9, 10, 12, 15, 95]`. Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.*  
*Hint: If the mean is pulled away from the median, the data is skewed in that direction.*

```python
import statistics

# Instagram follower counts (in thousands)
followers_k = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean_val = statistics.mean(followers_k)
median_val = statistics.median(followers_k)

print("=" * 65)
print("Task 2: Instagram Influencer Follower Analysis")
print("=" * 65)
print(f"Follower Counts (in '000s) : {followers_k}")
print(f"Calculated Mean (Average)  : {mean_val:.2f}k ({mean_val * 1000:,.0f} followers)")
print(f"Calculated Median (Middle) : {median_val:.2f}k ({median_val * 1000:,.0f} followers)\n")

print("=" * 65)
print("COMPARISON & SKEWNESS INFERENCE:")
print("=" * 65)
print(f"• Higher Metric: The MEAN ({mean_val:.2f}k) is significantly HIGHER than the MEDIAN ({median_val:.2f}k).")
print("• Skewness Determination: The data is RIGHT-SKEWED (Positively Skewed).")
print("• Reason: The single influencer with 95k followers acts as a strong right-tail outlier that drags the mean upwards to 18.78k, whereas the median remains at 9.00k, accurately reflecting typical micro-influencer sizes.")
```

## Task 3: Real-World Example of Right-Skewed Data

**Prompt:** *Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views). Briefly describe the dataset and explain why you think it is right skewed.*

---

### Selected Example: **YouTube Video Views**

#### 1. Dataset Description:
* **App / Platform:** YouTube
* **Data Variable:** Total view count per uploaded video across creator channels.
* **Observed Range:** From under 100 views on individual or amateur uploads to billions of views on global viral hits and music videos (e.g., *Baby Shark*, *Despacito*).

#### 2. Why it is Right-Skewed (Positively Skewed):
1. **Natural Lower Bound at Zero:** Video views cannot be negative ($Views \ge 0$). Over 90% of videos published on YouTube generate modest view counts (between 50 and 5,000 views). This creates a heavy, dense cluster of data points on the left side of the distribution.
2. **Infinite Upper Tail:** A very small fraction (top 0.1%) of videos go viral, accumulating millions or hundreds of millions of views. These extreme positive outliers stretch the tail far to the right.
3. **Statistical Relation ($\text{Mean} > \text{Median}$):** Because the huge view counts of viral videos heavily inflate the arithmetic average, the platform's **Mean view count** is significantly higher than the **Median view count** (which represents the typical creator experience). This rightward pull characterizes a classic **Right-Skewed Distribution**.

## Task 4: Flipkart Product Ratings - Graph & Skewness Justification

**Prompt:** *Given this dataset of Flipkart product ratings: `[3, 3, 4, 4, 4, 5, 5, 5, 5, 5]`, draw a rough graph and state if the distribution is symmetrical, left skewed, or right skewed. Justify your answer in one line.*

```python
import matplotlib.pyplot as plt
import statistics
from collections import Counter

# Flipkart product ratings dataset
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

mean_r = statistics.mean(ratings)
median_r = statistics.median(ratings)
mode_r = statistics.mode(ratings)
counts = Counter(ratings)

print("=" * 65)
print("Task 4: Flipkart Product Ratings Distribution")
print("=" * 65)
print(f"Ratings Dataset    : {ratings}")
print(f"Frequency Count    : {dict(counts)}")
print(f"Mean Rating        : {mean_r:.2f}")
print(f"Median Rating      : {median_r:.2f}")
print(f"Mode Rating        : {mode_r:.2f}\n")

# Plotting the Rating Distribution Bar Chart
stars = [3, 4, 5]
frequencies = [counts[s] for s in stars]

plt.figure(figsize=(7, 4.5))
bars = plt.bar(stars, frequencies, color=['#f39c12', '#2980b9', '#27ae60'], edgecolor='black', width=0.5)

# Add value labels above bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.1, f'{int(yval)}', ha='center', va='bottom', fontweight='bold')

plt.axvline(mean_r, color='red', linestyle='--', linewidth=2, label=f'Mean = {mean_r:.2f}')
plt.axvline(median_r, color='blue', linestyle='-', linewidth=2, label=f'Median = {median_r:.2f}')
plt.title("Flipkart Product Ratings (Left-Skewed Distribution)", fontsize=12, fontweight='bold')
plt.xlabel("Star Rating", fontsize=11)
plt.ylabel("Review Frequency (Count)", fontsize=11)
plt.xticks(stars, ['3 Stars', '4 Stars', '5 Stars'])
plt.ylim(0, 6)
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig("flipkart_ratings_distribution.png", dpi=150)
plt.show()

print("=" * 65)
print("CLASSIFICATION & ONE-LINE JUSTIFICATION:")
print("=" * 65)
print("• Distribution Classification: LEFT SKEWED (Negatively Skewed).")
print("• One-line Justification: The distribution is left-skewed because ratings are heavily clustered at the highest score of 5 stars with a long tail trailing towards lower values, resulting in Mean (4.30) < Median (4.50) < Mode (5.00).")
```
