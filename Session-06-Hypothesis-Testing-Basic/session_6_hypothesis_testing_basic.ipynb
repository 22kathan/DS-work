## Task 1: Hypotheses Formulation (Instagram vs. YouTube Daily Time)

**Prompt:** *Write down the null hypothesis (H0) and alternative hypothesis (H1) for the following scenario: You believe that people spend more time on Instagram than on YouTube daily.*

---

### Variable Definitions:
* Let $\mu_{\text{Instagram}}$ = Population mean daily time spent on Instagram (in minutes).
* Let $\mu_{\text{YouTube}}$ = Population mean daily time spent on YouTube (in minutes).

---

### Hypotheses:

#### 1. **Null Hypothesis ($H_0$):**
* **In Words:** There is no significant difference in the daily time spent on Instagram compared to YouTube, or people spend less than or equal time on Instagram compared to YouTube.
* **Mathematical Notation:**
  $$H_0: \mu_{\text{Instagram}} \le \mu_{\text{YouTube}} \quad \text{(or } \mu_{\text{Instagram}} - \mu_{\text{YouTube}} \le 0\text{)}$$
  *(In standard two-tailed testing: $H_0: \mu_{\text{Instagram}} = \mu_{\text{YouTube}}$)*

#### 2. **Alternative Hypothesis ($H_1$ / $H_a$):**
* **In Words:** People spend significantly more time on Instagram than on YouTube daily (Right-tailed / Directional claim).
* **Mathematical Notation:**
  $$H_1: \mu_{\text{Instagram}} > \mu_{\text{YouTube}} \quad \text{(or } \mu_{\text{Instagram}} - \mu_{\text{YouTube}} > 0\text{)}$$

## Task 2: Two-Sample t-Test on Zomato Home Page Redesign

**Prompt:** *Given the data: Group A (users shown a new Zomato home page) had an average order value of ₹350 from 30 users, and Group B (old home page) had an average order value of ₹320 from 30 users. Assume both groups have a standard deviation of ₹50. Calculate the p-value using a two-sample t-test (you may use Python, R, or an online calculator), and decide whether to accept or reject the null hypothesis at a 0.05 significance level.*  
*Hint: Use `scipy.stats.ttest_ind` in Python or any online t-test calculator if you are not comfortable with code.*

```python
import scipy.stats as stats
import numpy as np

# Group A: New Zomato Home Page
n_A = 30
mean_A = 350
std_A = 50

# Group B: Old Zomato Home Page
n_B = 30
mean_B = 320
std_B = 50

# Significance level
alpha = 0.05

# Two-sample t-test from sample summary statistics
t_stat, p_value_two_tailed = stats.ttest_ind_from_stats(
    mean1=mean_A, std1=std_A, nobs1=n_A,
    mean2=mean_B, std2=std_B, nobs2=n_B,
    equal_var=True
)

p_value_one_tailed = p_value_two_tailed / 2

# Degrees of freedom
df = n_A + n_B - 2

print("=" * 70)
print("Task 2: Two-Sample t-Test Results (Zomato A/B Test)")
print("=" * 70)
print(f"Group A (New Home Page) : Mean = ₹{mean_A}, Std = ₹{std_A}, n = {n_A}")
print(f"Group B (Old Home Page) : Mean = ₹{mean_B}, Std = ₹{std_B}, n = {n_B}")
print(f"Degrees of Freedom (df) : {df}")
print(f"Calculated t-statistic  : {t_stat:.4f}")
print(f"p-value (Two-tailed)    : {p_value_two_tailed:.4f}")
print(f"p-value (One-tailed)    : {p_value_one_tailed:.4f}")
print(f"Significance Level (α)  : {alpha}")

print("\n" + "=" * 70)
print("STATISTICAL DECISION & CONCLUSION:")
print("=" * 70)
if p_value_two_tailed < alpha:
    print(f"• Decision: REJECT the Null Hypothesis (H0) because p-value ({p_value_two_tailed:.4f}) < α ({alpha}).")
    print("• Conclusion: The difference in Average Order Value between the new home page and old home page is statistically significant.")
    print("• Business Insight: The new home page design genuinely increased order value by ₹30 on average; the increase is not due to random chance.")
else:
    print(f"• Decision: ACCEPT / FAIL TO REJECT the Null Hypothesis (H0) because p-value ({p_value_two_tailed:.4f}) >= α ({alpha}).")
```

## Task 3: A/B Test Design for a Real-World App Feature

**Prompt:** *Pick any feature from your favorite app (like the new dark mode in WhatsApp or a redesigned payment button in Paytm) and describe how you would set up an A/B test for it. Clearly state what your control and variant groups would be, what metric you would measure, and what your null and alternative hypotheses are.*

---

### Feature Selected: **"1-Tap Fast Checkout" Button on Paytm / Zomato**

#### 1. Background & Goal
Currently, completing a transaction requires 3 steps: confirming cart, selecting payment method, and confirming payment. The product team designed a sticky **"1-Tap UPI Fast Checkout"** button to eliminate friction.

---

### A/B Test Framework:

#### **A. Target Population & Allocation**
* **Population:** All active users initiating a checkout on the app during a 14-day test window.
* **Control Group (Group A - 50% traffic):** Users experience the existing multi-step checkout workflow.
* **Variant Group (Group B - 50% traffic):** Users are shown the new prominent "1-Tap UPI Fast Checkout" button pre-configured with their default UPI handle.

#### **B. Metrics to Measure**
* **Primary Metric (Conversion Rate):**
  $$\text{Checkout Conversion Rate} = \frac{\text{Number of Completed Transactions}}{\text{Number of Users Reaching Checkout Screen}} \times 100$$
* **Secondary / Guardrail Metrics:**
  * Average time to checkout (in seconds).
  * Transaction failure rate / customer support refund tickets (to ensure users don't make accidental payments).

#### **C. Hypotheses Formulation**
* **Null Hypothesis ($H_0$):**  
  The new "1-Tap Fast Checkout" button does not increase the checkout conversion rate compared to the existing design.
  $$H_0: p_{\text{variant}} \le p_{\text{control}}$$
* **Alternative Hypothesis ($H_1$):**  
  The new "1-Tap Fast Checkout" button significantly increases the checkout conversion rate.
  $$H_1: p_{\text{variant}} > p_{\text{control}}$$

#### **D. Decision Rule**
Run the experiment for 14 full days (accounting for weekday vs. weekend patterns) until sample size provides 80% statistical power at $\alpha = 0.05$. If $p\text{-value} < 0.05$, reject $H_0$ and roll out the 1-Tap button to 100% of users.

## Task 4: Flipkart Wishlist Button A/B Test Evaluation & p-Value Interpretation

**Prompt:** *Suppose you run an A/B test for a new 'Add to Wishlist' button design on Flipkart. Out of 1000 users, 520 clicked the new button versus 480 who clicked the old button. Without doing detailed calculations, explain if this difference is likely to be significant or just due to chance. Briefly explain how the p-value helps you decide.*

---

### Evaluation Without Detailed Calculations:

#### 1. **Is the Difference Likely Significant or Due to Chance?**
* **Conclusion:** The difference is **likely due to random chance (not statistically significant)**.
* **Reasoning:**
  * 520 clicks out of 1,000 represents a **52% click-through rate** for the new button versus **48% (480 clicks)** for the old button—a modest difference of only **4 percentage points** (or $\pm 2\%$ around a neutral 50-50 baseline).
  * In a sample of 1,000 users, standard binomial sampling variability naturally produces small fluctuations of $\pm 2\%$ to $\pm 3\%$ purely by chance. (This is analogous to flipping a fair coin 1,000 times and observing 520 Heads and 480 Tails; nobody would conclude the coin is biased based on this small difference).

---

### 2. **How the p-Value Helps You Decide:**

* **What the p-Value Represents:**  
  The $p$-value quantifies the probability of observing a difference of 40 clicks or larger purely by random luck, assuming both button designs are equally effective (the Null Hypothesis is true).
* **Decision Framework:**
  * If **$p\text{-value} < 0.05$**: The observed difference would be extremely rare under random noise alone ($\le 5\%$ probability). We would reject the null hypothesis and conclude the new button design genuinely causes higher engagement.
  * If **$p\text{-value} \ge 0.05$**: The observed difference is consistent with everyday random variation.
* **Application to this Test:**  
  A quick statistical calculation of this proportion difference yields a $p$-value of approximately **$0.20$ ($20\%$)**. Because $0.20 \gg 0.05$, there is a 1-in-5 chance that this 40-click gap occurred simply by random luck. Therefore, we **fail to reject the null hypothesis** and conclude the new design does not provide a statistically significant improvement.

```python
from scipy import stats

# Flipkart Wishlist Button Data
total_users = 1000
clicks_new = 520
clicks_old = 480

# Two-sided Binomial Test assuming neutral 50-50 chance (p=0.5)
res = stats.binomtest(clicks_new, n=total_users, p=0.5, alternative='two-sided')

print("=" * 65)
print("Task 4: Flipkart A/B Test Verification")
print("=" * 65)
print(f"Total Users Tested       : {total_users}")
print(f"Clicks on New Button     : {clicks_new} ({clicks_new/total_users:.1%})")
print(f"Clicks on Old Button     : {clicks_old} ({clicks_old/total_users:.1%})")
print(f"Calculated p-value       : {res.pvalue:.4f}")
print(f"Significance Level (α)   : 0.05\n")

if res.pvalue < 0.05:
    print("Result: Statistically Significant (p < 0.05)")
else:
    print(f"Result: NOT Statistically Significant (p = {res.pvalue:.4f} >= 0.05)")
    print("Conclusion: The 4% difference is within expected random chance variation.")
```
