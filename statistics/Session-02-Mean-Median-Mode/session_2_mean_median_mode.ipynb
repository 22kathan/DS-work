## Task 1: Mean Spotify Daily Play Counts

**Prompt:** *Create a Python script called `playlist_stats.py` that takes a list of daily Spotify song play counts (e.g., `[120, 135, 150, 200, 120, 90, 200]`) and calculates the mean (average) number of plays.*

```python
import statistics

# Daily song play counts from Spotify
daily_plays = [120, 135, 150, 200, 120, 90, 200]

# Function to calculate mean number of plays
def calculate_mean_plays(plays):
    return sum(plays) / len(plays)

mean_manual = calculate_mean_plays(daily_plays)
mean_builtin = statistics.mean(daily_plays)

print("=" * 50)
print("Task 1: Spotify Playlist Daily Play Statistics")
print("=" * 50)
print(f"Daily Play Counts : {daily_plays}")
print(f"Total Song Plays  : {sum(daily_plays)}")
print(f"Number of Days    : {len(daily_plays)}")
print(f"Mean Daily Plays  : {mean_manual:.2f} (Verified with statistics.mean: {mean_builtin:.2f})")
```

## Task 2: Median Delivery Time for Zomato Orders

**Prompt:** *Given the following array of delivery times (in minutes) for Zomato orders: `[30, 25, 40, 35, 30, 45, 30]`, write code to find the median delivery time and print it.*

```python
import statistics

# Array of delivery times (in minutes) for Zomato orders
delivery_times = [30, 25, 40, 35, 30, 45, 30]

# Finding the median delivery time
sorted_times = sorted(delivery_times)
median_time = statistics.median(delivery_times)

print("=" * 50)
print("Task 2: Zomato Order Delivery Times")
print("=" * 50)
print(f"Original Delivery Times : {delivery_times}")
print(f"Sorted Delivery Times   : {sorted_times}")
print(f"Middle Element (N=7)    : Index 3 -> {sorted_times[3]}")
print(f"Median Delivery Time    : {median_time} minutes")
```

## Task 3: Mode of Flipkart Product Ratings

**Prompt:** *Write a function `most_common_rating(ratings)` that takes a list of Flipkart product ratings (e.g., `[5, 4, 4, 3, 5, 4, 2, 4]`) and returns the mode (most frequent rating).*

```python
import statistics
from collections import Counter

def most_common_rating(ratings):
    """
    Takes a list of Flipkart product ratings and returns the mode (most frequent rating).
    """
    return statistics.mode(ratings)

# Sample list of Flipkart product ratings
flipkart_ratings = [5, 4, 4, 3, 5, 4, 2, 4]

# Function execution
mode_result = most_common_rating(flipkart_ratings)
rating_counts = Counter(flipkart_ratings)

print("=" * 50)
print("Task 3: Flipkart Product Ratings Analysis")
print("=" * 50)
print(f"Product Ratings List : {flipkart_ratings}")
print("Rating Frequencies:")
for star, count in sorted(rating_counts.items(), key=lambda x: x[0], reverse=True):
    print(f"  • {star} Stars: {count} times")

print(f"\n=> Mode (Most Common Rating): {mode_result} Stars (appeared {rating_counts[mode_result]} times)")
```

## Task 4: Comparing Mean, Median & Mode on YouTube Channel Views (Outlier Sensitivity)

**Prompt:** *Given three lists representing YouTube video views for three different channels, compare the mean, median, and mode for each channel, and decide which channel's data is most affected by outliers.*  
*Hint: Try using one list with a very high value (e.g., `[100, 120, 110, 105, 5000]`) to see the effect on mean vs median.*

```python
import statistics

# Views for 3 different YouTube channels
channel_A = [100, 120, 110, 105, 115]        # Channel A: Steady, consistent views
channel_B = [100, 120, 110, 105, 5000]       # Channel B: Has extreme outlier (viral video)
channel_C = [200, 220, 250, 220, 210]        # Channel C: Normal distribution with repeating value

channels = {
    "Channel A (Consistent)": channel_A,
    "Channel B (Viral Outlier)": channel_B,
    "Channel C (Steady Moderate)": channel_C
}

print("=" * 80)
print("Task 4: YouTube Views Comparison & Outlier Sensitivity")
print("=" * 80)
print(f"{'Channel':<28} | {'Mean':<10} | {'Median':<10} | {'Mode':<14} | {'Outlier Impact'}")
print("-" * 80)

for name, views in channels.items():
    mean_val = statistics.mean(views)
    median_val = statistics.median(views)
    try:
        mode_val = str(statistics.mode(views))
    except statistics.StatisticsError:
        mode_val = "No unique mode"
        
    diff = abs(mean_val - median_val)
    impact = f"Extreme Skew (Diff={diff:.1f})" if diff > 100 else f"Low (Diff={diff:.1f})"
    print(f"{name:<28} | {mean_val:<10.1f} | {median_val:<10.1f} | {mode_val:<14} | {impact}")

print("\n" + "=" * 80)
print("DECISION & REASONING:")
print("=" * 80)
print("• Channel B's data is overwhelmingly affected by the outlier (5,000 views).")
print("• The single viral video pulls Channel B's Mean up to 1087.0, whereas the Median remains stable at 110.0.")
print("• Conclusion: The Mean is sensitive to extreme values, while the Median is robust against outliers.")
```

## Task 5: Real-World Dataset Analysis with Pandas (Kaggle Swiggy Restaurants)

**Prompt:** *Pick any real dataset from Kaggle or UCI (e.g., IPL player runs, Swiggy restaurant ratings, or movie box office collections), load it using pandas, and display the mean, median, and mode for one numeric column of your choice.*

```python
import pandas as pd

# Load dataset (Swiggy restaurant data from Kaggle)
df = pd.read_csv("swiggy_restaurants.csv")

print("=" * 60)
print("Task 5: Real-World Dataset Preview (Kaggle Swiggy Data)")
print("=" * 60)
print(df[['restaurant_name', 'cuisine', 'rating', 'cost_for_two']].head())

# Selected numeric column: 'cost_for_two'
col = "cost_for_two"

mean_cost = df[col].mean()
median_cost = df[col].median()
mode_cost = df[col].mode()[0]

print("\n" + "=" * 60)
print(f"Statistical Summary for '{col}':")
print("=" * 60)
print(f"• Mean   (Average Cost) : ₹{mean_cost:.2f}")
print(f"• Median (Middle Cost)  : ₹{median_cost:.2f}")
print(f"• Mode   (Common Cost)  : ₹{mode_cost:.2f}")

# Also displaying for 'rating' column
print("\n" + "=" * 60)
print("Statistical Summary for 'rating':")
print("=" * 60)
print(f"• Mean   (Average Rating): {df['rating'].mean():.2f}")
print(f"• Median (Middle Rating) : {df['rating'].median():.2f}")
print(f"• Mode   (Common Rating) : {df['rating'].mode()[0]:.2f}")
```
