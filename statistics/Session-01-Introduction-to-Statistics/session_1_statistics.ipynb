## Task 2: Classification of Music App Sample Data

**Prompt:** *Given this sample data from a music app: `['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']`, classify the type of data and explain your reasoning in 2-3 lines.*

---

### Classification:
* **Data Type:** **Categorical (Qualitative) Data**
* **Sub-Type:** **Nominal Data**

### Reasoning (2–3 lines):
> This dataset comprises discrete text labels representing music genres with no natural numerical magnitude, hierarchy, or mathematical ranking (e.g., *Pop* is neither mathematically larger nor smaller than *Jazz*). Arithmetic calculations such as addition or calculating a mathematical mean cannot be performed; only frequency distribution counts, percentages, and the **Mode** (`'Pop'`, frequency = 3) are statistically meaningful.

```python
from collections import Counter
import statistics

# Task 2 Sample Data
music_data = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']

# Frequency calculation
genre_counts = Counter(music_data)
mode_genre = statistics.mode(music_data)

print("=" * 55)
print("Task 2: Sample Music App Data Analysis")
print("=" * 55)
print(f"Sample Dataset: {music_data}\n")
print("Genre Frequencies & Proportions:")
for genre, count in genre_counts.items():
    pct = (count / len(music_data)) * 100
    print(f"  • {genre:<8} : {count} plays ({pct:.1f}%)")

print(f"\n=> Mode (Most Frequently Occurring Genre): '{mode_genre}' ({genre_counts[mode_genre]} occurrences)")
```

## Task 3: Enhancing App Features Using Statistics

**Prompt:** *Pick any one feature from an app you use (such as the number of likes on an Instagram post, or the rating of a restaurant on Zomato) and describe how statistics could help improve that feature for users.*

---

### Selected Feature: **Restaurant Rating System on Zomato**

#### The Problem with Simple Averages:
1. **Low Sample-Size Distortion:** A brand-new restaurant with 1 review of 5.0 stars currently outranks an established restaurant with a 4.7-star rating across 2,000 reviews.
2. **Susceptibility to Extreme Outliers:** A few malicious 1-star reviews ("review bombing") or fake 5-star reviews disproportionately distort arithmetic means for lower-volume restaurants.

#### How Statistical Methods Solve This:
1. **Bayesian Weighted Average (Empirical Bayes):**  
   Instead of displaying the raw sample mean $R$, compute a smoothed Bayesian average $W$:
   $$W = \frac{v \cdot R + m \cdot C}{v + m}$$
   *where:*
   - $v$ = number of reviews for the restaurant
   - $R$ = average rating of the restaurant
   - $m$ = minimum review threshold (confidence weighting factor, e.g., 50 reviews)
   - $C$ = global platform-wide average rating (e.g., 4.0 stars)
   
   *Result:* Restaurants with few reviews are pulled toward the platform baseline ($C$). As reviews ($v$) grow, the rating naturally approaches its true sample average ($R$).

2. **Interquartile Range (IQR) & Trimmed Mean:**  
   By calculating the **Interquartile Range** ($Q_3 - Q_1$) and trimming values beyond $1.5 \times \text{IQR}$, automated bot attacks or anomalous 1-star review spikes are neutralized before computing the average.

3. **Confidence Interval Display:**  
   Displaying a confidence band (e.g., *"4.6 ± 0.1 stars based on 1,500+ orders"*) communicates statistical reliability directly to the user.

```python
def calculate_bayesian_rating(raw_avg, review_count, global_avg=4.0, min_threshold=50):
    """Calculates smoothed Bayesian weighted average rating."""
    return (review_count * raw_avg + min_threshold * global_avg) / (review_count + min_threshold)

restaurants = [
    {"name": "New Bistro (Low Review Volume)", "raw_avg": 5.0, "reviews": 1},
    {"name": "Famous Diner (High Review Volume)", "raw_avg": 4.7, "reviews": 1500},
    {"name": "Mid-tier Cafe", "raw_avg": 4.2, "reviews": 60},
    {"name": "Targeted Pizzeria (Review Bombed)", "raw_avg": 2.5, "reviews": 4}
]

print("=" * 78)
print("Task 3: Bayesian Smoothing Simulation vs. Raw Average")
print("=" * 78)
print(f"{'Restaurant Name':<35} | {'Reviews':<8} | {'Raw Avg':<10} | {'Bayesian Rating':<16}")
print("-" * 78)

for r in restaurants:
    bayes_score = calculate_bayesian_rating(r["raw_avg"], r["reviews"])
    print(f"{r['name']:<35} | {r['reviews']:<8} | {r['raw_avg']:<10.1f} | {bayes_score:<16.2f}")
```

## Task 4: Statistical Terms for Playlist Recommendations (Spotify-like App)

**Prompt:** *Suppose you are building a new playlist recommendation feature for a Spotify-like app. List 3 basic statistical terms (such as mean, median, mode, variance, etc.) that would be useful in analyzing user listening data, and explain each with a one-line example.*

---

### Selected Statistical Metrics:

1. **Mode (Most Frequent Value):**
   * **Definition & Application:** Identifies the categorical attribute that appears most often in a user's listening stream (e.g., top genre, dominant artist, or preferred language).
   * **One-Line Example:** *Detecting that a listener's most played genre between 8:00 AM and 11:00 AM is "Lo-Fi Beats" to automatically generate a personalized morning study playlist.*

2. **Mean (Arithmetic Average):**
   * **Definition & Application:** Calculates the baseline central value of continuous numerical acoustic attributes (such as Track Tempo in Beats Per Minute [BPM], Danceability score, or Track Duration).
   * **One-Line Example:** *Computing a user's average track tempo of 128 BPM during running sessions to queue workout songs matching that specific cadence.*

3. **Variance / Standard Deviation (Measure of Dispersion):**
   * **Definition & Application:** Quantifies how concentrated or widely scattered a user's acoustic preferences are around their mean listening profile.
   * **One-Line Example:** *A low variance in track acousticness and energy indicates consistent listening tastes (ideal for ambient playlists), whereas high variance signals an eclectic taste profile suited for "Discover Weekly" mix exploration.*

```python
# Sample listening metrics for a user's workout session
listening_bpm = [124, 128, 130, 128, 132, 135, 128, 126, 130]
listening_genres = ['EDM', 'Workout Pop', 'EDM', 'Rock', 'EDM', 'Hip-Hop']

mean_tempo = statistics.mean(listening_bpm)
median_tempo = statistics.median(listening_bpm)
mode_genre = statistics.mode(listening_genres)
var_tempo = statistics.variance(listening_bpm)
std_tempo = statistics.stdev(listening_bpm)

print("=" * 60)
print("Task 4: Playlist Statistics (Mean, Mode, Variance)")
print("=" * 60)
print(f"Recorded Track Tempos (BPM) : {listening_bpm}")
print(f"Listening Genres History    : {listening_genres}\n")
print(f"1. Mean Track Tempo        : {mean_tempo:.1f} BPM (Median: {median_tempo:.1f} BPM)")
print(f"2. Mode Preferred Genre    : '{mode_genre}'")
print(f"3. Variance in Tempo       : {var_tempo:.2f} (Standard Deviation: {std_tempo:.2f} BPM)")
```
