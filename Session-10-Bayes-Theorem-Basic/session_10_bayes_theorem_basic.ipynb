## Task 1: General Bayes' Posterior Probability Function

**Prompt:** *Write a Python function `bayes_posterior(prior, likelihood, evidence)` that calculates and returns the posterior probability using Bayes theorem formula for given values.*

---

### Bayes' Theorem Formula:
$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$
*where:*
* **$P(A)$ (Prior):** Initial probability of hypothesis $A$ before seeing evidence $B$.
* **$P(B \mid A)$ (Likelihood):** Probability that evidence $B$ occurs given that hypothesis $A$ is true.
* **$P(B)$ (Evidence / Total Probability):** Total marginal probability of observing evidence $B$.
* **$P(A \mid B)$ (Posterior):** Updated probability of hypothesis $A$ given that evidence $B$ has been observed.

```python
def bayes_posterior(prior, likelihood, evidence):
    """
    Calculates the posterior probability P(A|B) using Bayes' Theorem.
    
    Parameters:
        prior (float): P(A) - Probability of hypothesis before evidence
        likelihood (float): P(B|A) - Probability of evidence given hypothesis
        evidence (float): P(B) - Total probability of the evidence
        
    Returns:
        float: P(A|B) - Updated posterior probability
    """
    if evidence <= 0:
        raise ValueError("Evidence P(B) must be strictly greater than 0.")
    if not (0 <= prior <= 1 and 0 <= likelihood <= 1 and 0 <= evidence <= 1):
        raise ValueError("All probability inputs must be bounded between 0 and 1.")
        
    posterior = (likelihood * prior) / evidence
    return posterior

# Example Test: Medical diagnostic test
# Prior disease prevalence: 1% (0.01)
# Test Sensitivity (Likelihood): 95% (0.95)
# Total positive test rate (Evidence): 5.9% (0.059)
test_prior = 0.01
test_likelihood = 0.95
test_evidence = 0.059

result = bayes_posterior(test_prior, test_likelihood, test_evidence)

print("=" * 65)
print("Task 1: Bayes' Theorem Function Demonstration")
print("=" * 65)
print(f"Prior Probability P(A)           : {test_prior:.4f}")
print(f"Likelihood P(B|A)                : {test_likelihood:.4f}")
print(f"Evidence P(B)                    : {test_evidence:.4f}")
print(f"Calculated Posterior P(A|B)      : {result:.4f} ({result * 100:.2f}%)")
```

## Task 2: Swiggy Premium Restaurant Discount Scenario

**Prompt:** *Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, while only 20% of regular restaurants do. If you see a discount, use Bayes theorem in code to find the probability that the restaurant is premium.*  
*Hint: Define prior, likelihood, and evidence in your code with comments.*

---

### Mathematical Breakdown:
* **Events:**
  * Let $P = \text{Restaurant is Premium}$, and $R = \text{Restaurant is Regular}$.
  * Let $D = \text{Restaurant offers a Discount}$.
* **Given Probabilities:**
  * **Prior:** $P(P) = 0.10 \implies P(R) = 1 - 0.10 = 0.90$
  * **Likelihood for Premium:** $P(D \mid P) = 0.80$
  * **Likelihood for Regular:** $P(D \mid R) = 0.20$
* **Total Probability of Evidence $P(D)$:**
  $$P(D) = P(D \mid P) \cdot P(P) + P(D \mid R) \cdot P(R) = (0.80 \times 0.10) + (0.20 \times 0.90) = 0.08 + 0.18 = 0.26$$
* **Posterior Probability $P(P \mid D)$:**
  $$P(P \mid D) = \frac{P(D \mid P) \cdot P(P)}{P(D)} = \frac{0.80 \times 0.10}{0.26} = \frac{0.08}{0.26} \approx \mathbf{0.3077\ (30.77\%)}$$

```python
# ==========================================
# Task 2: Swiggy Premium Discount Bayes Model
# ==========================================

# 1. PRIOR PROBABILITY:
# P(Premium): Baseline probability that a restaurant chosen at random is premium
prior_premium = 0.10

# P(Regular): Probability that a restaurant is regular (non-premium)
prior_regular = 1.0 - prior_premium  # 0.90

# 2. LIKELIHOOD:
# P(Discount | Premium): Probability a premium restaurant offers a discount
likelihood_discount_given_premium = 0.80

# P(Discount | Regular): Probability a regular restaurant offers a discount
likelihood_discount_given_regular = 0.20

# 3. EVIDENCE (TOTAL PROBABILITY OF OBSERVING A DISCOUNT):
# Using the Law of Total Probability:
# P(Discount) = P(Discount|Premium)*P(Premium) + P(Discount|Regular)*P(Regular)
evidence_discount = (likelihood_discount_given_premium * prior_premium) + \
                    (likelihood_discount_given_regular * prior_regular)

# 4. POSTERIOR PROBABILITY:
# P(Premium | Discount) using Bayes' Theorem:
posterior_premium_given_discount = (likelihood_discount_given_premium * prior_premium) / evidence_discount

print("=" * 65)
print("Task 2: Swiggy Restaurant Bayes Theorem Results")
print("=" * 65)
print(f"1. Prior P(Premium)                        : {prior_premium:.2%}")
print(f"   Prior P(Regular)                        : {prior_regular:.2%}")
print(f"2. Likelihood P(Discount | Premium)        : {likelihood_discount_given_premium:.2%}")
print(f"   Likelihood P(Discount | Regular)        : {likelihood_discount_given_regular:.2%}")
print(f"3. Total Evidence P(Discount)              : {evidence_discount:.4f} ({evidence_discount:.2%})")
print(f"\n=> 4. Posterior P(Premium | Discount)      : {posterior_premium_given_discount:.4f} ({posterior_premium_given_discount:.2%})")

print("\n" + "=" * 65)
print("INTERPRETATION:")
print("=" * 65)
print("• Initially, there was only a 10% chance that a random restaurant is premium.")
print("• Observing that it offers a discount updates our belief, raising the probability to 30.77% (a 3x increase).")
print("• It is still under 50% because regular restaurants make up 90% of the platform, so even their 20% discount rate produces more total discounts (18%) than premium restaurants (8%).")
```

## Task 3: Naive Bayes Email Spam Filter Simulation

**Prompt:** *Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.*

```python
# Dataset of 20 email subjects with binary label: 1 = Spam, 0 = Ham (Not Spam)
emails = [
    ("Claim your free gift card now", 1),
    ("Free trial ends tonight act fast", 1),
    ("Meeting agenda for Monday morning", 0),
    ("Win a free cruise vacation today", 1),
    ("Your project review report is ready", 0),
    ("Exclusive free entry to the event", 1),
    ("Can we reschedule our lunch tomorrow", 0),
    ("Invoice attached for your recent payment", 0),
    ("Get free coins on your next recharge", 1),
    ("Important security update for your account", 0),
    ("Congratulations you won a free coupon", 1),
    ("Notes from today team standup meeting", 0),
    ("Flight ticket confirmation from airline", 0),
    ("Join us for a free webinar on data science", 0),  # Non-spam with 'free'
    ("Doctor appointment scheduled for Friday", 0),
    ("Quarterly financial earnings report", 0),
    ("Urgent: Claim 100% free bonus cash", 1),
    ("Your order has been shipped and is out for delivery", 0),
    ("Weekly newsletter and engineering blog", 0),
    ("Reminder: Submit your timesheet by 5pm", 0)
]

total_emails = len(emails)
spam_emails = [text for text, label in emails if label == 1]
ham_emails = [text for text, label in emails if label == 0]

# 1. Priors:
p_spam = len(spam_emails) / total_emails
p_ham = len(ham_emails) / total_emails

# 2. Likelihoods: Check presence of word 'free' (case-insensitive)
target_word = "free"

spam_with_word = sum(1 for text in spam_emails if target_word in text.lower())
ham_with_word = sum(1 for text in ham_emails if target_word in text.lower())

p_word_given_spam = spam_with_word / len(spam_emails)
p_word_given_ham = ham_with_word / len(ham_emails)

# 3. Evidence (Total probability of word 'free'):
p_word = (p_word_given_spam * p_spam) + (p_word_given_ham * p_ham)

# 4. Posterior P(Spam | 'free'):
p_spam_given_word = (p_word_given_spam * p_spam) / p_word

print("=" * 70)
print(f"Task 3: Spam Detection Filter Simulation (Word: '{target_word}')")
print("=" * 70)
print(f"Total Emails Analyzed         : {total_emails}")
print(f"  • Spam Count                : {len(spam_emails)} ({p_spam:.1%})")
print(f"  • Ham (Legitimate) Count    : {len(ham_emails)} ({p_ham:.1%})")
print(f"\nLikelihoods:")
print(f"  • P('{target_word}' | Spam)           : {spam_with_word}/{len(spam_emails)} = {p_word_given_spam:.2%}")
print(f"  • P('{target_word}' | Ham)            : {ham_with_word}/{len(ham_emails)} = {p_word_given_ham:.2%}")
print(f"\nTotal Evidence P('{target_word}')       : {p_word:.4f} ({p_word:.1%})")
print(f"\n=> Posterior P(Spam | '{target_word}')   : {p_spam_given_word:.4f} ({p_spam_given_word:.2%})")

print("\n" + "=" * 70)
print("FILTER DECISION:")
print("=" * 70)
print(f"• If an email contains '{target_word}', the probability that it is SPAM surges from 35.0% (prior) to {p_spam_given_word:.1%} (posterior).")
print(f"• Since {p_spam_given_word:.1%} exceeds standard 80% filtering thresholds, the email would be routed directly to the Spam folder.")
```

### 2. The Generated Scenario (Fake Review Detection on Flipkart):
* **Context:** An e-commerce platform like Flipkart employs a machine learning model to detect fake or sponsored bot reviews.
* **Given Numbers:**
  1. **Prior:** $4\%$ of all submitted product reviews are fake/bot-generated ($P(\text{Fake}) = 0.04$, meaning $P(\text{Genuine}) = 0.96$).
  2. **Model Sensitivity (Likelihood):** If a review is indeed fake, the model correctly flags it $92\%$ of the time ($P(\text{Flagged} \mid \text{Fake}) = 0.92$).
  3. **False Positive Rate:** If a review is genuine, the model mistakenly flags it $3\%$ of the time ($P(\text{Flagged} \mid \text{Genuine}) = 0.03$).
* **Problem to Solve:** If a review is flagged by the algorithm, what is the actual posterior probability that it is truly fake ($P(\text{Fake} \mid \text{Flagged})$)?

```python
# ==========================================
# Task 4: Flipkart Fake Review Detection
# ==========================================

# 1. Priors:
prior_fake = 0.04       # P(Fake) = 4%
prior_genuine = 0.96    # P(Genuine) = 96%

# 2. Likelihoods:
likelihood_flagged_given_fake = 0.92      # Sensitivity: P(Flagged | Fake)
likelihood_flagged_given_genuine = 0.03   # False Positive Rate: P(Flagged | Genuine)

# 3. Evidence: Total Probability of a review being flagged
# P(Flagged) = P(Flagged|Fake)*P(Fake) + P(Flagged|Genuine)*P(Genuine)
evidence_flagged = (likelihood_flagged_given_fake * prior_fake) + \
                   (likelihood_flagged_given_genuine * prior_genuine)

# 4. Posterior Probability: P(Fake | Flagged)
posterior_fake_given_flagged = (likelihood_flagged_given_fake * prior_fake) / evidence_flagged

print("=" * 70)
print("Task 4: Flipkart Fake Review Detection (Bayes Theorem)")
print("=" * 70)
print(f"Prior Probability P(Fake)                        : {prior_fake:.2%}")
print(f"Prior Probability P(Genuine)                     : {prior_genuine:.2%}")
print(f"Likelihood P(Flagged | Fake)                     : {likelihood_flagged_given_fake:.2%}")
print(f"Likelihood P(Flagged | Genuine) [False Positive] : {likelihood_flagged_given_genuine:.2%}")
print(f"\nTotal Evidence P(Flagged)                        : {evidence_flagged:.4f} ({evidence_flagged:.2%})")
print(f"\n=> Posterior P(Fake | Flagged)                   : {posterior_fake_given_flagged:.4f} ({posterior_fake_given_flagged:.2%})")

print("\n" + "=" * 70)
print("KEY DATA SCIENCE INSIGHT (Base Rate Fallacy):")
print("=" * 70)
print(f"• Despite the model's high 92% detection rate, a flagged review has only a {posterior_fake_given_flagged:.1%} chance of being fake.")
print("• Why? Because genuine reviews are so overwhelmingly common (96%), the 3% false alarm rate produces almost as many false flags (0.0288) as true flags (0.0368).")
print("• Product Recommendation: Do not auto-delete flagged reviews; instead, route them to human moderation or prompt the buyer for verified purchase proof.")
```
