## Task 1: Step-by-Step Manual Variance Calculation for Fitness Steps

**Prompt:** *Given an array of daily step counts from your fitness app for one week (e.g., `[4500, 7000, 5000, 8000, 4000, 9000, 6000]`), calculate the variance by hand using the variance formula and show each calculation step.*

---

### Step-by-Step Manual Calculation

#### **Step 1: Identify Data & Count**
* Data points: $X = [4500, 7000, 5000, 8000, 4000, 9000, 6000]$
* Number of observations ($n$ or $N$): **7**

#### **Step 2: Calculate the Mean ($\bar{x}$)**
$$\bar{x} = \frac{\sum x_i}{n} = \frac{4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000}{7} = \frac{43500}{7} \approx 6214.29 \text{ steps}$$

#### **Step 3: Calculate Deviations and Squared Deviations**

| Day ($i$) | Steps ($x_i$) | Deviation $(x_i - \bar{x})$ | Squared Deviation $(x_i - \bar{x})^2$ |
|---|---|---|---|
| Day 1 | 4,500 | $4500 - 6214.29 = -1714.29$ | $2,938,775.51$ |
| Day 2 | 7,000 | $7000 - 6214.29 = +785.71$ | $617,346.94$ |
| Day 3 | 5,000 | $5000 - 6214.29 = -1214.29$ | $1,474,489.80$ |
| Day 4 | 8,000 | $8000 - 6214.29 = +1785.71$ | $3,188,775.51$ |
| Day 5 | 4,000 | $4000 - 6214.29 = -2214.29$ | $4,903,061.22$ |
| Day 6 | 9,000 | $9000 - 6214.29 = +2785.71$ | $7,760,204.08$ |
| Day 7 | 6,000 | $6000 - 6214.29 = -214.29$ | $45,918.37$ |
| **Total** | $\sum x_i = 43,500$ | $\sum (x_i - \bar{x}) = 0$ | **$\sum (x_i - \bar{x})^2 = 20,928,571.43$** |

#### **Step 4: Compute the Variance**
* **Sample Variance ($s^2$, $n-1 = 6$):**
  $$s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1} = \frac{20,928,571.43}{6} \approx \mathbf{3,488,095.24}$$

* **Population Variance ($\sigma^2$, $N = 7$):**
  $$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N} = \frac{20,928,571.43}{7} \approx \mathbf{2,989,795.92}$$

```python
import statistics

# Step counts for one week
step_counts = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

n = len(step_counts)
mean_steps = sum(step_counts) / n

# Sum of squared deviations
squared_diffs = [(x - mean_steps) ** 2 for x in step_counts]
sum_squared_diffs = sum(squared_diffs)

sample_variance = sum_squared_diffs / (n - 1)
pop_variance = sum_squared_diffs / n

print("=" * 65)
print("Task 1: Fitness Step Counts Variance Calculation")
print("=" * 65)
print(f"Step Counts Dataset        : {step_counts}")
print(f"Number of Days (n)         : {n}")
print(f"Mean Steps (x̄)             : {mean_steps:.2f}")
print(f"Sum of Squared Diff Σ(x-x̄)² : {sum_squared_diffs:,.2f}")
print(f"Sample Variance (n-1 = 6)  : {sample_variance:,.2f} (Verified: {statistics.variance(step_counts):,.2f})")
print(f"Population Variance (N = 7): {pop_variance:,.2f} (Verified: {statistics.pvariance(step_counts):,.2f})")
```

## Task 2: Python Function for Standard Deviation

**Prompt:** *Write a Python function called `calculate_standard_deviation(scores)` that takes a list of exam scores and returns the standard deviation rounded to two decimal places.*

```python
import statistics

def calculate_standard_deviation(scores):
    """
    Takes a list of exam scores and returns the sample standard deviation
    rounded to two decimal places.
    """
    if len(scores) < 2:
        return 0.00
    
    mean_val = sum(scores) / len(scores)
    variance_val = sum((x - mean_val) ** 2 for x in scores) / (len(scores) - 1)
    std_dev = variance_val ** 0.5
    return round(std_dev, 2)

# Test cases with student exam scores
exam_scores_1 = [85, 90, 78, 92, 88, 76, 95]
exam_scores_2 = [70, 72, 68, 71, 69]
exam_scores_3 = [50, 95, 60, 90, 45, 100]

print("=" * 60)
print("Task 2: Testing calculate_standard_deviation(scores)")
print("=" * 60)
print(f"Exam Scores Batch 1 : {exam_scores_1}")
print(f"Standard Deviation  : {calculate_standard_deviation(exam_scores_1)} (Verified: {round(statistics.stdev(exam_scores_1), 2)})\n")

print(f"Exam Scores Batch 2 : {exam_scores_2}")
print(f"Standard Deviation  : {calculate_standard_deviation(exam_scores_2)} (Verified: {round(statistics.stdev(exam_scores_2), 2)})\n")

print(f"Exam Scores Batch 3 : {exam_scores_3}")
print(f"Standard Deviation  : {calculate_standard_deviation(exam_scores_3)} (Verified: {round(statistics.stdev(exam_scores_3), 2)})")
```

## Task 3: Comparing Daily Zomato Spending (Consistency Analysis)

**Prompt:** *Imagine two friends track their daily spending on Zomato over 5 days. Friend A spends `[200, 200, 200, 200, 200]` and Friend B spends `[100, 300, 150, 400, 50]`. Calculate the standard deviation for both and explain in 2-3 lines who has more consistent spending and why.*

```python
import statistics

# Spending over 5 days (in ₹)
friend_A = [200, 200, 200, 200, 200]
friend_B = [100, 300, 150, 400, 50]

mean_A = statistics.mean(friend_A)
mean_B = statistics.mean(friend_B)

stdev_A = statistics.stdev(friend_A)
stdev_B = statistics.stdev(friend_B)

print("=" * 65)
print("Task 3: Daily Zomato Spending Comparison")
print("=" * 65)
print(f"Friend A Daily Spending : {friend_A}")
print(f"Friend A Mean Spending  : ₹{mean_A:.2f}")
print(f"Friend A Standard Dev   : ₹{stdev_A:.2f}\n")

print(f"Friend B Daily Spending : {friend_B}")
print(f"Friend B Mean Spending  : ₹{mean_B:.2f}")
print(f"Friend B Standard Dev   : ₹{stdev_B:.2f}\n")

print("=" * 65)
print("CONSISTENCY EXPLANATION (2-3 lines):")
print("=" * 65)
print("• Friend A has vastly more consistent spending with a Standard Deviation of ₹0.00, meaning their daily spending never deviates from the average of ₹200.")
print("• In contrast, Friend B exhibits high spending volatility (Standard Deviation = ₹145.77), swinging drastically from ₹50 to ₹400 across the 5 days despite having the exact same ₹200 mean.")
```

## Task 4: Startup Monthly Salary Variance & Spread Interpretation

**Prompt:** *You are given the monthly salary data of 6 employees in a startup: `[28000, 29000, 27000, 60000, 26500, 27500]`. Calculate the variance and standard deviation, then interpret in 2-3 lines what the high or low spread means for this team.*

```python
import statistics

# Startup employee monthly salaries (in ₹)
salaries = [28000, 29000, 27000, 60000, 26500, 27500]

n = len(salaries)
mean_sal = statistics.mean(salaries)
median_sal = statistics.median(salaries)
sample_var = statistics.variance(salaries)
sample_std = statistics.stdev(salaries)

print("=" * 70)
print("Task 4: Startup Monthly Salary Variance & Spread Analysis")
print("=" * 70)
print(f"Salaries Dataset       : {salaries}")
print(f"Number of Employees    : {n}")
print(f"Mean Salary            : ₹{mean_sal:,.2f}")
print(f"Median Salary          : ₹{median_sal:,.2f}")
print(f"Sample Variance        : {sample_var:,.2f}")
print(f"Sample Standard Dev    : ₹{sample_std:,.2f}\n")

print("=" * 70)
print("SPREAD INTERPRETATION (2-3 lines):")
print("=" * 70)
print("• The startup team has an exceptionally high spread (Standard Deviation = ₹13,255.19 relative to a ₹33,000 mean).")
print("• Five of the employees earn very tightly clustered salaries between ₹26,500 and ₹29,000, while a single outlier (₹60,000, likely a founder or lead) heavily skews the variance.")
print("• This indicates significant internal compensation disparity and highlights why median (₹27,750) better represents typical team pay than the mean.")
```
