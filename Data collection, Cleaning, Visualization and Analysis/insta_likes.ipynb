## Task 2: Instagram Influencer Follower Counts (`np.array`)

**Task Description:**
> *Create a NumPy array called `followers` using `np.array()` that stores the follower counts for 5 Instagram influencers: `[1200, 15000, 67000, 340000, 1250000]`. Print the array, its shape, number of dimensions, and data type.*

### Core Attributes Explained:
- **`shape`**: A tuple of integers indicating the size of the array along each dimension. For a 1D array of 5 elements, shape is `(5,)`.
- **`ndim`**: The number of axes (dimensions) of the array. A flat sequence has `ndim = 1`.
- **`dtype`**: The data type of the elements stored in the array. Default for 64-bit platforms is `int64`.
- **`size`**: Total number of elements across all dimensions ($5$).

```python
# Task 2: Create followers array and inspect its attributes
followers = np.array([1200, 15000, 67000, 340000, 1250000])

print("Instagram Followers Array :", followers)
print("Array Shape               :", followers.shape)
print("Number of Dimensions (ndim):", followers.ndim)
print("Data Type (dtype)         :", followers.dtype)
print("Total Elements (size)     :", followers.size)
```

## Task 3: Sequential Zomato Order IDs (`np.arange`)

**Task Description:**
> *Use `np.arange()` to generate an array of order IDs for 10 consecutive Zomato orders starting from 101. Print the array and its size.*

### `np.arange()` Mechanics:
- Syntax: `np.arange(start, stop, step)`
- Like Python's built-in `range()`, the `stop` boundary is exclusive: $[\text{start}, \text{stop})$.
- To generate 10 consecutive order IDs from $101$, the sequence ends at $110$, requiring `stop = 111` (`np.arange(101, 111)`).

```python
# Task 3: Generate 10 consecutive Zomato order IDs starting at 101
zomato_order_ids = np.arange(101, 111)

print("Zomato Consecutive Order IDs:", zomato_order_ids)
print("Total Orders Generated (size):", zomato_order_ids.size)
print("Start ID                     :", zomato_order_ids[0])
print("End ID                       :", zomato_order_ids[-1])
```

## Task 4: Spotify Playlist 'Like' Identity Matrix (`np.eye`)

**Task Description:**
> *Create a 3x3 NumPy array using `np.eye()` to represent a 'like' identity matrix for a new Spotify playlist feature. Print the matrix and explain what the diagonal values represent in a comment.*

### Identity Matrix in Recommendation & Graph Systems:
In recommendation engines and network graphs:
- An **Adjacency / Interaction Matrix** $M$ has rows and columns corresponding to entities (tracks, playlists, or users).
- **Diagonal Elements ($M_{i,i} = 1.0$)**: Represent **self-similarity** or **reflexive affinity**—every track or playlist naturally has a 100% affinity/identical match with itself.
- **Off-Diagonal Elements ($M_{i,j} = 0.0, i \neq j$)**: Represent the initial baseline state before cross-track co-listening or user playlist interactions are factored in.

```python
# Task 4: Create 3x3 Spotify 'like' identity matrix
spotify_like_matrix = np.eye(3)

# COMMENT EXPLANATION:
# In this Spotify playlist affinity matrix:
# - Diagonal values (matrix[i, i] == 1.0) represent 100% self-similarity or direct self-affinity
#   (a playlist or track compared to itself is an exact match / guaranteed 'like').
# - Off-diagonal elements (0.0) indicate that distinct tracks have zero default cross-similarity
#   prior to calculating recommendation weights or shared listener metrics.

print("Spotify 'Like' Identity Matrix (3x3):\n", spotify_like_matrix)
print("\nDiagonal Values                     :", spotify_like_matrix.diagonal())
print("Off-Diagonal Elements Are Zero?     :", np.all(spotify_like_matrix[~np.eye(3, dtype=bool)] == 0))
```

## Task 5: Cricket Scores List Conversion & Memory Footprint (`.itemsize`)

**Task Description:**
> *Convert a Python list of cricket scores `[45, 67, 120, 89, 54]` to a NumPy array, then use the `.itemsize` attribute to print how many bytes each score takes in memory.*
> *Hint: Use `np.array()` for conversion and `.itemsize` for memory size.*

### Contiguous Memory Layout vs. Python Lists:
- **Python List (`list`)**: An array of pointers (references) pointing to disparate `PyObject` integers scattered throughout the heap. Each integer incurs an object header overhead (typically 28 bytes per integer in 64-bit Python) plus an 8-byte pointer reference.
- **NumPy `ndarray`**: A single contiguous block of raw binary data (e.g. 64-bit integer = 8 bytes).
  - `.itemsize`: Returns the size in bytes of each individual element.
  - `.nbytes`: Returns the total bytes consumed by all elements (`size * itemsize`).

```python
# Task 5: Convert cricket scores and inspect memory footprint
cricket_scores_list = [45, 67, 120, 89, 54]
cricket_scores_array = np.array(cricket_scores_list)

print("Original Python List        :", cricket_scores_list)
print("Converted NumPy Array       :", cricket_scores_array)
print("Array Data Type (dtype)     :", cricket_scores_array.dtype)
print("Bytes per Score (itemsize)  :", cricket_scores_array.itemsize, "bytes")
print("Total Buffer Memory (nbytes):", cricket_scores_array.nbytes, "bytes")
print("Is Memory C-Contiguous?     :", cricket_scores_array.flags["C_CONTIGUOUS"])
```

## Supplementary Concepts: Array Creation Family

To ensure comprehensive mastery over all topics listed in **Session 1**, here is a synthesis of the additional array generation utilities:
- `np.zeros(shape)`: Pre-allocating zero matrices for gradients or masks.
- `np.ones(shape)`: Initializing baseline weights or scale multipliers.
- `np.full(shape, fill_value)`: Initializing matrices with a uniform default value.
- `np.linspace(start, stop, num)`: Generating evenly spaced intervals for continuous plotting and loss landscapes.

```python
# Supplementary Array Creation Examples
zeros_demo = np.zeros((2, 3), dtype=float)
ones_demo = np.ones((2, 3), dtype=int)
full_demo = np.full((2, 3), 99)
linspace_demo = np.linspace(0, 100, 5)

print("np.zeros((2, 3))     :\n", zeros_demo)
print("\nnp.ones((2, 3))      :\n", ones_demo)
print("\nnp.full((2, 3), 99)  :\n", full_demo)
print("\nnp.linspace(0, 100, 5):", linspace_demo)
```
