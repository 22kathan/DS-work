## Task 1: Instagram Hours Spent vs. Posts Made (Scatter Plot & Visual Analysis)

**Prompt:** *Create two Python lists: one representing the number of hours you spend on Instagram each day for a week, and another for the number of posts you make each day. Plot these as a scatter plot using matplotlib and visually describe whether the relationship looks like a positive, negative, or zero correlation.*

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for 7 days (one week)
# Hours spent on Instagram each day
hours_spent = [1.5, 2.0, 3.5, 1.0, 4.0, 2.5, 3.0]

# Number of posts/stories made each day
posts_made = [1, 2, 3, 0, 4, 2, 3]

# Create Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(hours_spent, posts_made, color='#e1306c', s=100, edgecolors='black', linewidth=1.5, label='Daily Activity')

# Add trend line
m, b = np.polyfit(hours_spent, posts_made, 1)
x_line = np.linspace(min(hours_spent), max(hours_spent), 50)
plt.plot(x_line, m * x_line + b, color='#833ab4', linestyle='--', linewidth=2, label='Linear Trend')

plt.title("Instagram Daily Usage: Hours Spent vs. Posts Made", fontsize=12, fontweight='bold')
plt.xlabel("Hours Spent on Instagram per Day", fontsize=11)
plt.ylabel("Number of Posts Made per Day", fontsize=11)
plt.yticks(range(0, 6))
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig("instagram_scatter_plot.png", dpi=150)
plt.show()

print("=" * 70)
print("Task 1: Visual Relationship & Correlation Description")
print("=" * 70)
print(f"Daily Hours Spent : {hours_spent}")
print(f"Daily Posts Made  : {posts_made}")
print("\nVISUAL CORRELATION DESCRIPTION:")
print("• The scatter plot shows a clear upward trend from bottom-left to top-right.")
print("• As the number of hours spent on Instagram increases, the number of posts created also consistently rises.")
print("• Therefore, the relationship visually exhibits a strong POSITIVE CORRELATION (r > 0).")
```

## Task 2: Monthly Zomato vs. Swiggy Order Count Correlation (`numpy.corrcoef`)

**Prompt:** *Given the following data for 6 friends: their monthly Zomato order count and their monthly Swiggy order count, calculate the correlation coefficient between the two lists using numpy's corrcoef(). Print the result and explain if the correlation is positive, negative, or zero.*

```python
import numpy as np

# Monthly order count data for 6 friends
friends = ["Friend 1", "Friend 2", "Friend 3", "Friend 4", "Friend 5", "Friend 6"]
zomato_orders = [15, 12, 20, 8, 25, 10]
swiggy_orders = [14, 11, 19, 7, 24, 9]

# Calculate Pearson correlation matrix using numpy's corrcoef()
correlation_matrix = np.corrcoef(zomato_orders, swiggy_orders)
r = correlation_matrix[0, 1]

print("=" * 65)
print("Task 2: Correlation Between Zomato & Swiggy Orders")
print("=" * 65)
print(f"{'Friend':<12} | {'Zomato Orders':<15} | {'Swiggy Orders':<15}")
print("-" * 48)
for f, z, s in zip(friends, zomato_orders, swiggy_orders):
    print(f"{f:<12} | {z:<15} | {s:<15}")

print("\n" + "=" * 65)
print("Numpy Correlation Matrix:")
print(correlation_matrix)
print(f"\nPearson Correlation Coefficient (r) : {r:.4f}")

print("\n" + "=" * 65)
print("EXPLANATION:")
print("=" * 65)
print("• Type of Correlation: STRONG POSITIVE CORRELATION (r = +0.9984 ≈ +1.00).")
print("• Interpretation: The correlation coefficient is extremely close to +1.0, showing a nearly perfect direct linear relationship.")
print("• Meaning: Friends who frequently order food on Zomato are also heavy users of Swiggy, indicating high general reliance on food delivery services rather than an exclusive preference for one platform over the other.")
```

## Task 4: Strong Negative Correlation Dataset (Spotify Time vs. Test Marks)

**Prompt:** *Use ChatGPT to generate a small dataset (10 pairs of numbers) that shows a strong negative correlation between time spent on Spotify and marks scored in a test. Paste the dataset and plot it as a scatter plot using matplotlib.*  
*Hint: Ask ChatGPT to "generate 10 pairs of numbers with a strong negative correlation" and specify your variables.*

```python
import matplotlib.pyplot as plt
import numpy as np

# 10 pairs of data generated showing strong negative correlation
# Variable X: Daily hours spent on Spotify
spotify_hours = [0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0]

# Variable Y: Test marks scored (out of 100)
test_marks    = [94,  89,  85,  80,  76,  71,  66,  60,  54,  48]

# Calculate Pearson correlation coefficient
r_negative = np.corrcoef(spotify_hours, test_marks)[0, 1]

print("=" * 65)
print("Task 4: Dataset Table (Spotify Hours vs. Test Marks)")
print("=" * 65)
print(f"{'Student #':<12} | {'Spotify Hours (X)':<20} | {'Test Marks (Y)':<15}")
print("-" * 52)
for i, (h, m_score) in enumerate(zip(spotify_hours, test_marks), start=1):
    print(f"Student {i:<4} | {h:<20.1f} | {m_score:<15}")

print(f"\nPearson Correlation Coefficient (r): {r_negative:.4f}")
print("Classification: STRONG NEGATIVE CORRELATION (r ≈ -0.9996)\n")

# Scatter Plot with Negative Trendline
plt.figure(figsize=(7.5, 4.5))
plt.scatter(spotify_hours, test_marks, color='#1db954', s=100, edgecolors='black', linewidth=1.5, label='Student Data Points')

# Fit linear regression trend line
slope, intercept = np.polyfit(spotify_hours, test_marks, 1)
x_vals = np.linspace(min(spotify_hours), max(spotify_hours), 50)
plt.plot(x_vals, slope * x_vals + intercept, color='#e74c3c', linestyle='--', linewidth=2, label=f'Trendline (r = {r_negative:.2f})')

plt.title("Study Week: Spotify Listening Hours vs. Test Marks", fontsize=12, fontweight='bold')
plt.xlabel("Daily Time Spent on Spotify (Hours)", fontsize=11)
plt.ylabel("Marks Scored in Test (out of 100)", fontsize=11)
plt.ylim(40, 100)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig("spotify_vs_marks_negative_correlation.png", dpi=150)
plt.show()
```
