## Task 1: 1D Vector - Spotify Playlist Durations & Total Playtime

**Prompt:** *Create a Python list to represent a playlist of 5 song durations in seconds (e.g., `[210, 180, 245, 200, 175]`). Print the total duration of the playlist.*

```python
# 1D Vector (List) representing 5 song durations in seconds
song_durations = [210, 180, 245, 200, 175]

# Calculate total duration
total_seconds = sum(song_durations)

# Format into minutes and seconds
total_minutes = total_seconds // 60
remaining_seconds = total_seconds % 60

print("=" * 60)
print("Task 1: Spotify Playlist Duration Vector")
print("=" * 60)
print(f"Song Durations Vector (seconds) : {song_durations}")
print(f"Number of Songs in Playlist     : {len(song_durations)}")
print(f"Total Duration (Seconds)        : {total_seconds} seconds")
print(f"Total Duration (Formatted)      : {total_minutes} minutes, {remaining_seconds} seconds ({total_seconds / 60:.2f} mins)")
```

## Task 2: 3x3 Matrix - Swiggy Food Item Ratings & Slicing

**Prompt:** *Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends to 3 different food items on Swiggy. Print the matrix and display the rating each friend gave to the second food item.*

```python
import numpy as np

# Friends and Food Items metadata
friends = ["Friend 1 (Aman)", "Friend 2 (Priya)", "Friend 3 (Rohan)"]
food_items = ["Item 1 (Biryani)", "Item 2 (Paneer Pizza)", "Item 3 (Burger)"]

# 3x3 Matrix representing ratings from 1 to 5
# Rows = Friends, Columns = Food Items
ratings_matrix = np.array([
    [5, 4, 3],  # Friend 1 ratings
    [4, 5, 2],  # Friend 2 ratings
    [3, 4, 5]   # Friend 3 ratings
])

print("=" * 65)
print("Task 2: Swiggy Food Ratings Matrix (3x3)")
print("=" * 65)
print("Full 3x3 Ratings Matrix:")
print(ratings_matrix)

print("\nTabular Matrix Representation:")
print(f"{'Friend':<20} | {'Biryani':<10} | {'Paneer Pizza':<12} | {'Burger':<8}")
print("-" * 58)
for i, friend in enumerate(friends):
    print(f"{friend:<20} | {ratings_matrix[i, 0]:<10} | {ratings_matrix[i, 1]:<12} | {ratings_matrix[i, 2]:<8}")

# Slicing the second food item (Column Index 1)
# Column 1 corresponds to "Item 2 (Paneer Pizza)"
second_item_ratings = ratings_matrix[:, 1]

print("\n" + "=" * 65)
print(f"Ratings for the Second Food Item [{food_items[1]}]:")
print("=" * 65)
for friend, rating in zip(friends, second_item_ratings):
    print(f"• {friend:<20} gave : {rating} / 5 Stars")
```

## Task 3: 3x4 Matrix - Dish Orders Across Days & Column Aggregation

**Prompt:** *Given a matrix representing the number of orders for 4 different dishes across 3 days (rows: days, columns: dishes), write code to calculate and print the total orders for each dish.*

```python
import numpy as np

# Metadata for days and dishes
days = ["Day 1 (Friday)", "Day 2 (Saturday)", "Day 3 (Sunday)"]
dishes = ["Hyderabadi Biryani", "Paneer Butter Masala", "Masala Dosa", "Butter Chicken"]

# Matrix: 3 Days (Rows) x 4 Dishes (Columns)
orders_matrix = np.array([
    [25, 18, 40, 15],  # Day 1 orders
    [30, 22, 35, 20],  # Day 2 orders
    [28, 25, 45, 18]   # Day 3 orders
])

# Column-wise summation (axis=0 sums across rows down each column)
total_orders_per_dish = np.sum(orders_matrix, axis=0)

print("=" * 70)
print("Task 3: Daily Dish Orders Matrix (3 Days x 4 Dishes)")
print("=" * 70)
print(f"{'Day':<18} | {'Biryani':<10} | {'Paneer':<10} | {'Dosa':<10} | {'Chicken':<10}")
print("-" * 65)
for i, day in enumerate(days):
    print(f"{day:<18} | {orders_matrix[i, 0]:<10} | {orders_matrix[i, 1]:<10} | {orders_matrix[i, 2]:<10} | {orders_matrix[i, 3]:<10}")

print("\n" + "=" * 70)
print("TOTAL ORDERS FOR EACH DISH (Summed Across All 3 Days):")
print("=" * 70)
for dish, total in zip(dishes, total_orders_per_dish):
    print(f"• {dish:<24} : {total} total orders")

print(f"\nGrand Total Orders Across All Dishes: {np.sum(orders_matrix)} orders")
```

## Task 4: 5x2 Product Matrix - Price & Quantity Sold (Revenue Function)

**Prompt:** *Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. Represent this data as a matrix in Python, then write a function to calculate the total revenue for each product.*  
*Hint: Multiply price and quantity for each row, then print the results.*

```python
import numpy as np

# Product names
products = [
    "Wireless Bluetooth Earbuds",
    "Shockproof Smartphone Case",
    "Fast Charging Power Bank",
    "Braided USB-C Cable",
    "Fitness Smartwatch"
]

# 5 Products x 2 Columns Matrix: [Unit Price (₹), Quantity Sold]
flipkart_sales_matrix = np.array([
    [1499, 120],  # Earbuds: ₹1,499 each, 120 sold
    [299,  350],  # Case: ₹299 each, 350 sold
    [1199,  85],  # Power Bank: ₹1,199 each, 85 sold
    [199,  500],  # Cable: ₹199 each, 500 sold
    [2499,  60]   # Smartwatch: ₹2,499 each, 60 sold
])

def calculate_product_revenues(sales_matrix, product_names=None):
    """
    Calculates total revenue for each product from a [Price, Quantity] matrix.
    Revenue = Price * Quantity (Row-wise element-wise multiplication)
    """
    prices = sales_matrix[:, 0]
    quantities = sales_matrix[:, 1]
    
    # Vectorized element-wise multiplication
    revenues = prices * quantities
    return revenues

# Execute function
product_revenues = calculate_product_revenues(flipkart_sales_matrix, products)
total_store_revenue = np.sum(product_revenues)

print("=" * 75)
print("Task 4: Flipkart Product Sales Matrix & Revenue Breakdown")
print("=" * 75)
print(f"{'Product Name':<30} | {'Price (₹)':<10} | {'Quantity':<10} | {'Total Revenue (₹)':<18}")
print("-" * 75)

for i in range(len(products)):
    price = flipkart_sales_matrix[i, 0]
    qty = flipkart_sales_matrix[i, 1]
    rev = product_revenues[i]
    print(f"{products[i]:<30} | ₹{price:<9,d} | {qty:<10,d} | ₹{rev:<17,d}")

print("=" * 75)
print(f"Grand Total Flipkart Sales Revenue: ₹{total_store_revenue:,.2f}")
```
