## Task 2: Python Random Sampling Simulation

**Prompt:** *Write a Python script that simulates random sampling: given a list of 1000 fake user IDs (e.g., `user1` to `user1000`), select 50 random users to simulate a survey sample. Print the selected user IDs.*  
*Hint: Use the `random.sample()` function from Python's random module.*

```python
import random

# Step 1: Generate a population of 1000 fake user IDs
population_users = [f"user{i}" for i in range(1, 1001)]

# Step 2: Use random.sample() to select 50 random users without replacement
# Setting a seed for reproducible demonstration
random.seed(42)
sample_size = 50
selected_users = random.sample(population_users, sample_size)

print("=" * 70)
print(f"Task 2: Random Sampling Simulation (Selected {sample_size} Users from {len(population_users)})")
print("=" * 70)
print(f"Total Population Size : {len(population_users)} users")
print(f"Sample Size Drawn     : {len(selected_users)} users\n")
print("Selected Survey Sample User IDs:")

# Display the 50 sampled users in clean columns
for i in range(0, len(selected_users), 5):
    row = selected_users[i:i+5]
    print("  " + "  ".join(f"{u:<12}" for u in row))

print("\n" + "=" * 70)
print(f"Verification: Are all 50 selected users unique? -> {len(set(selected_users)) == 50}")
```

## Task 3: Why Use Sampling Instead of Full Population Data (Swiggy Delivery Times)

**Prompt:** *Imagine you are analyzing food delivery times on Swiggy. Explain in 2-3 sentences why you would use sampling instead of collecting data from every single order ever made.*

---

### Explanation:
> 1. **Resource & Computational Efficiency:** Swiggy handles hundreds of millions of historical orders across years; extracting and aggregating every single record would require massive cloud compute and database querying costs, slowing down operations.
> 2. **Statistical Sufficiency & Real-time Actionability:** According to the **Central Limit Theorem**, a well-drawn random sample of several thousand recent orders provides virtually the exact same statistical precision (tight confidence intervals with minimal margin of error) in seconds, allowing analysts to detect delivery bottlenecks and optimize routing algorithms in real time.

## Task 4: Identifying Bias in Music Preference Survey (Spotify Scenario)

**Prompt:** *Given this scenario: A survey about music preferences is only sent to users who recently streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.*

---

### Bias Analysis:

* **Type of Bias:** **Sampling Bias** (specifically **Selection Bias** / **Undercoverage Bias**).

* **Detailed Explanation:**
  > 1. **Non-Representative Sample:** The sampling frame was restricted exclusively to individuals who recently engaged with Bollywood tracks, systematically ignoring all users whose primary listening consists of International Pop, EDM, Rock, Classical, Hip-Hop, or regional Indian music (e.g., Punjabi, Tamil, Telugu).
  > 2. **Distorted Findings:** Because the surveyed group already possesses a demonstrated affinity for Bollywood, the survey results will severely overestimate the popularity, streaming hours, and positive sentiment toward Bollywood music across Spotify as a whole.
  > 3. **Violation of Equal Probability:** For an unbiased estimation of general music preferences, every active Spotify subscriber should have an equal, non-zero probability of being selected. Filtering by a specific genre prior to sampling creates an unrepresentative sample that cannot generalize to Spotify's broader user population.

## Task 5: Sampling Methods to Study BookMyShow Movie Booking Habits

**Prompt:** *Choose any two types of sampling (for example: random sampling, stratified sampling, systematic sampling, cluster sampling). For each, describe how you would use it to select a sample of users from BookMyShow to study movie booking habits.*

---

### Selected Sampling Method 1: **Stratified Sampling**

* **Concept:** Dividing the population into distinct, non-overlapping subgroups (strata) based on a meaningful characteristic, then drawing proportional random samples from each stratum.
* **How to Apply to BookMyShow:**
  1. **Identify Strata:** Segment the BookMyShow user base by **Booking Frequency** into three tiers:
     * *Frequent Moviegoers:* $\ge 3$ tickets per month (e.g., 15% of user base).
     * *Occasional Moviegoers:* 1 ticket every 1–2 months (e.g., 55% of user base).
     * *Rare / Event-driven Moviegoers:* Only on festivals/blockbusters like Diwali or Marvel releases (e.g., 30% of user base).
  2. **Sample Selection:** For a target sample of 1,000 users, randomly draw 150 frequent users, 550 occasional users, and 300 rare users.
* **Advantage:** Guarantees that low-frequency and high-frequency users are both represented in their exact true proportions, preventing dominant user groups from skewing the findings.

---

### Selected Sampling Method 2: **Systematic Sampling**

* **Concept:** Selecting elements at regular, fixed intervals through an ordered list after a random starting point.
* **How to Apply to BookMyShow:**
  1. **Order the Population:** Sequence all transactions or booking confirmation IDs recorded over a blockbuster opening weekend (e.g., $N = 100,000$ bookings arranged chronologically).
  2. **Determine Sampling Interval ($k$):** For a desired sample of $n = 1,000$ users:
     $$k = \frac{N}{n} = \frac{100,000}{1,000} = 100$$
  3. **Select Users:** Randomly choose a starting number between 1 and 100 (e.g., booking #27), then select every $100^{\text{th}}$ transaction thereafter (e.g., bookings #27, #127, #227, #327, ..., #99,927).
* **Advantage:** Extremely simple and computationally efficient to implement in a real-time booking stream, ensuring an even temporal spread across peak Friday night, Saturday matinee, and Sunday evening showtimes.
